## 1. Quick recap: functions and reusability

A function is a reusable block of code. Instead of writing the same logic repeatedly, define it once and call it whenever required.

The instructor revisits function composition: one function can call another function and use the value returned by it. The supplied notebook uses a scientific-looking formula as an example:
1. `cube_and_add(a, b)` returns the sum of the cubes.
2. `inverse(c)` returns the reciprocal.
3. `formula(a, b)` calls both functions in sequence.
4. `formula_v2()` and `formula_v3()` show that the nested calls can be written more compactly.

The mathematical expression is only an example to demonstrate function reuse and composition.

````python
def cube_and_add(a, b):
    return a**3 + b**3

def inverse(c):
    return c**-1

def formula(a, b):
    print("calculation started")
    value_1 = cube_and_add(a, b)
    value_2 = inverse(value_1)
    return value_2

def formula_v2(a, b):
    print("calculation started")
    value_1 = inverse(cube_and_add(a, b))
    return value_1

def formula_v3(a, b):
    print("calculation started")
    return inverse(cube_and_add(a, b))

print(formula(1, 2))
print(formula_v2(1, 2))
print(formula_v3(1, 2))
````

### 1.1 Breaking down the nested call

For `formula_v3(1, 2)`:
- `cube_and_add(1, 2)` calculates \(1^3 + 2^3 = 9\).
- `inverse(9)` calculates \(9^{-1} = 1/9\).
- `formula_v3()` returns that result to the caller.

The same calculation can be written using intermediate variables or by nesting one function call inside another. Intermediate variables can make a longer calculation easier to inspect and debug.

````python
# Step-by-step form
value_1 = cube_and_add(1, 2)
value_2 = inverse(value_1)
print(value_2)

# Nested form
print(inverse(cube_and_add(1, 2)))
````

## 2. The four broad function combinations

The instructor summarizes functions using two questions:
1. Does the function accept input parameters?
2. Does the function return a value?

This gives four combinations:

| Function form | Parameters? | Explicit return? |
|---|---|---|
| Body only | No | No |
| Body + return | No | Yes |
| Body + parameters | Yes | No |
| Body + parameters + return | Yes | Yes |

These are conceptual combinations, not four separate Python keywords. The function body contains the work; parameters receive input; `return` sends a result back to the caller.

````python
# 1. Body only: no parameters, no explicit return
def show_message():
    print("Hello")

# 2. Body + return: no parameters, returns a value
def get_message():
    return "Hello"

# 3. Body + parameters: accepts input, prints result, no explicit return
def show_addition(a, b):
    print(a + b)

# 4. Body + parameters + return
def get_addition(a, b):
    return a + b

show_message()
print(get_message())
show_addition(10, 20)
answer = get_addition(10, 20)
print(answer)
````

### 2.1 Parameters and arguments

The instructor uses the idea of a function having “hands” to receive inputs. The technical definition is:

- **Parameters** are placeholders written in the function definition.
- **Arguments** are the actual values supplied when the function is called.

Example:

```python
def add(a, b):
    return a + b

answer = add(1, 2)
```

Here, `a` and `b` are parameters. The values `1` and `2` are arguments. When the call runs, Python binds `a = 1` and `b = 2`.

````python
def add(a, b):
    return a + b

answer = add(1, 2)
print(answer)
````

## 4. Default arguments

A default argument gives a parameter a value in the function definition. If the caller omits that argument, Python uses the default.

The class example uses `a=0`. The reason is that adding zero does not change the other value, so the function can still produce a useful result when the caller omits `a`.

````python
def add(a=0):
    print(a)
    print(a + 10)

add()       # Uses the default a=0
add(100)    # Uses the supplied a=100
````

### 4.1 Multiple default parameters

Both parameters can have defaults. Then the function can be called with no arguments, one argument or two arguments.

````python
def add(a=0, b=0):
    return a + b

print(add())
print(add(10))
print(add(10, 20))
````

### 4.2 Required parameters and default parameters

The class demonstrates this valid pattern:

```python
def add(a, b=0):
    return a + b
```

Parameter `a` is required, while `b` has a default. A required parameter must not be placed after a parameter with a default in this ordinary positional/keyword parameter section. For example, `def add(a=0, b):` is invalid syntax.

````python
def add(a, b=0):
    return a + b

print(add(10))
print(add(10, 20))

# Invalid definition — do not run:
# def invalid_add(a=0, b):
#     return a + b
````

## 5. Arbitrary positional arguments — `*args`

Sometimes a function should accept a variable number of positional arguments. You may not know in advance whether the caller will supply zero, one, two or many values.

The class introduces the asterisk (`*`) before a parameter name for this purpose:

```python
def add(*value):
    ...
```

The name `value` is not special. You can use another valid variable name. The important syntax is the single asterisk.

Python collects the extra positional arguments into a **tuple**.

````python
def add(*value):
    print(type(value))
    print(value)

add()
add(1, 2)
add("hello", 1, 2)
add("hello", 1, 2, [1, 2, 3])
add("hello", 1, 2, [1, 2, 3], {1: "hello"}, (1, 2, 3))
````

### 5.1 Why is `*value` a tuple?

The instructor explains that `*value` catches the positional values as tuple elements:
- No extra values → empty tuple `()`.
- One extra value → one-element tuple.
- Multiple extra values → tuple containing all of them.

Because a tuple is iterable, you can loop over the collected values.

````python
def show_values(*value):
    for item in value:
        print(item)

show_values(1, 2, 3)
````

### 5.2 Add any number of numeric values

The supplied notebook builds `add_v3(*value)`. It starts a total at zero, loops over the tuple and adds each value to the total.

````python
def add_v3(*value):
    total = 0
    for i in value:
        total += i
    return total

print("We are adding values:", add_v3(1, 2, 3, 4, 5, 67, 567, 4.50))
````

**Note:** this example expects numeric arguments. If a string, dictionary or list is passed to this sum function, `total += i` may raise a `TypeError`. `*args` accepts many kinds of positional values, but the function's own logic still determines which types it can process.

### 5.3 Combining named parameters with `*args`

The notebook next defines `a` and `b` as named parameters, then uses `*value` to collect the remaining positional arguments.

````python
def add_v3(a=0, b=0, *value):
    print(f"a value : {a}")
    print(f"b value : {b}")
    print(f"value value : {value}")

add_v3(1, 2, 3, 4, 5, 6, 7, 8)
````

For the call above:
- `a` receives `1`.
- `b` receives `2`.
- `value` receives `(3, 4, 5, 6, 7, 8)`.

The first two positional arguments are assigned to the named parameters. The remaining positional arguments are collected in the tuple.

## 6. Positional arguments

A positional argument is matched to a parameter based on its position in the function call.

For this function:

```python
def add_v4(a, b, c):
    ...
```

the first positional argument goes to `a`, the second to `b`, and the third to `c`.

````python
def add_v4(a, b, c):
    print(a, b, c)

add_v4(1, 2, 3)
````

## 7. Keyword arguments

A keyword argument specifies the parameter name explicitly. This lets the caller provide values by name rather than relying only on their order.

The class example calls the function with the keyword names in a different order:

````python
def add_v4(a, b, c):
    print(a, b, c)

add_v4(c=2, b=45, a=100)
````

### 7.1 Mixing positional and keyword arguments

The supplied notebook also demonstrates positional arguments followed by a keyword argument:

````python
def add_v4(a, b, c):
    print(a, b, c)

add_v4(1, 2, c=4)
````

The positional values are assigned first (`a=1`, `b=2`), and `c=4` is assigned by its keyword.

**Rule demonstrated in class:** positional arguments should come before keyword arguments in a function call. Do not place a positional argument after a keyword argument.

````python
# Valid: positional arguments first, then keyword argument
add_v4(1, 2, c=4)

# Invalid call — do not run:
# add_v4(a=1, 2, 3)
````

## 8. Required keyword-only parameters after `*args`

The notebook introduces this signature:

```python
def add_v4(a, b, *values, c):
    ...
```

After `*values`, further ordinary parameters such as `c` are keyword-only. The caller must provide `c` by name. Any extra positional arguments are collected inside `values`.

````python
def add_v4(a, b, *values, c):
    print("a:", a)
    print("b:", b)
    print("values:", values)
    print("c:", c)

add_v4(1, 2, 3, 4, c=5)
````

The original class notebook also contains this call:

```python

```

It does not match the function signature: `d` and `e` are not accepted parameters, and `f` is not defined. This is useful as an example of why the function signature and the call must agree. To accept additional named arguments, the next section introduces `**new_value`.

## 9. Arbitrary keyword arguments — `**kwargs`

A double asterisk before a parameter collects extra keyword arguments into a **dictionary**.

The class uses the name `new_value`, and mentions that developers commonly call this parameter `kwargs`. The name is not special; the double asterisk is the important syntax.

```python
def add_v5(a, b, *values, **new_value):
    ...
```

- `a` and `b` receive the first two positional arguments.
- `*values` collects additional positional arguments into a tuple.
- `**new_value` collects additional keyword arguments into a dictionary.

````python
def add_v5(a, b, *values, **new_value):
    print("a:", a)
    print("b:", b)
    print("values:", values)
    print("new_value:", new_value)

add_v5(
    1, 2, 3, 4, 5, 6, 7, 8,
    c=2,
    batch="ultimate data science 3.0"
)
````

### 9.1 `*` versus `**`

| Syntax | Collects | Resulting type |
|---|---|---|
| `*values` | Extra positional arguments | Tuple |
| `**new_value` | Extra keyword arguments | Dictionary |

The class's shorthand recap is:
- `*` → tuple
- `**` → dictionary

Remember that the function definition uses these markers to collect arguments. They are not interchangeable.

````python
def inspect_arguments(a, b, *t, **d):
    print("a:", a)
    print("b:", b)
    print("extra positional:", t, type(t))
    print("extra keyword:", d, type(d))

inspect_arguments(1, 2, 3, 4, 5, 6, 7, {"c": 2}, ab=2)
````

### 9.2 Invalid ordering example in the original notebook

The supplied notebook contains:

```python
add_v5(1, 2, 3, 4, 5, 6, 7, 8, c=2, batch="ultimate data science 3.0", 9)
```

This call is invalid because the positional argument `9` appears after keyword arguments. Keep positional arguments first, then keyword arguments.

## 10. Flexible functions: a calculator driven by keyword arguments

The instructor uses `**action_todo` to create a function that can perform one or more requested operations. The caller selects operations by passing keyword names such as `add=True` or `subtract=True`.

The function:
1. Receives `a` and `b`.
2. Collects requested operations in a dictionary called `action_todo`.
3. Creates an empty result dictionary called `answer`.
4. Checks which operation names are present.
5. Calculates each requested operation and stores it in `answer`.
6. Returns the dictionary.

The class uses separate `if` statements rather than `elif`, because the caller may request more than one operation in a single call.

````python
def calculation(a, b, **action_todo):
    answer = {}
    check_action = list(action_todo.keys())

    if "add" in check_action:
        answer["add"] = a + b

    if "subtract" in check_action:
        answer["subtract"] = a - b

    if "multiply" in check_action:
        answer["multiply"] = a * b

    if "divide" in check_action:
        answer["divide"] = a / b

    return answer

print(calculation(10, 20, add=True, subtract=True))
print(calculation(10, 20, divide=True, subtract=True))
print(calculation(10, 20, multiply=True, subtract=True, c=3))
````

### 10.1 Dry run: `calculation(10, 20, add=True, subtract=True)`

The function receives:
- `a = 10`
- `b = 20`
- `action_todo = {"add": True, "subtract": True}`

It checks the dictionary's keys:
- `"add"` is present → store `30`.
- `"subtract"` is present → store `-10`.
- `"multiply"` is not present → skip.
- `"divide"` is not present → skip.

The returned dictionary is:

```python
{"add": 30, "subtract": -10}
```

### Why use multiple `if` statements?
If the function used `if` / `elif`, only the first matching operation would run. Independent `if` statements allow several requested operations to be calculated in one call.

### 10.2 Safer variation for division

The original class example directly divides `a / b`. If `b` is zero, Python raises `ZeroDivisionError`. This small practice variation checks for zero before dividing. It is an additional defensive improvement, not a separate topic taught in the class.

````python
def calculation_safe(a, b, **action_todo):
    answer = {}

    if "add" in action_todo:
        answer["add"] = a + b

    if "subtract" in action_todo:
        answer["subtract"] = a - b

    if "multiply" in action_todo:
        answer["multiply"] = a * b

    if "divide" in action_todo:
        if b == 0:
            answer["divide"] = "Cannot divide by zero"
        else:
            answer["divide"] = a / b

    return answer

print(calculation_safe(10, 0, add=True, divide=True))
````

## 11. The built-in `print()` function also accepts keyword arguments

The instructor uses `print()` to demonstrate positional arguments and keyword arguments. The values to be printed are positional arguments; `sep` and `end` are keyword arguments that control how the output is formatted.

### 11.1 `sep` — separator between values

By default, `print()` separates multiple values with a space.

````python
print("123", "hello", 123)
````

You can change that separator using `sep`. For example, `sep=","` places a comma between the values.

````python
print("123", "hello", 123, sep=",")
print("123", "hello", 123, sep=" | ")
print("hello", "world", sep="-")
````

### 11.2 `end` — what is printed after the values

By default, `print()` ends with a newline character (`\n`). That is why consecutive `print()` calls normally appear on separate lines.

You can change this behavior with `end`.

````python
print("hello", end="\n")
print("world", end="---------\n")
print("123")
````

If you use a tab character (`\t`), the next output begins after a tab space. If you use an empty string as `end`, the next print output continues on the same line.

````python
print("hello", end="\t")
print("world", end="---------\n")
print("123")
````

````python
print("hello", end="")
print("world", end="---------")
print("123")
````

### 11.3 Using `sep` and `end` together

The class demonstrates that `sep` controls separation **between values in one `print()` call**, while `end` controls what is added **after that call has printed all its values**.

````python
print("hello", "world", "12345", sep="#", end="---------\n")
print("123")
````

### 11.4 Common escape sequences shown in class

| Sequence | Meaning |
|---|---|
| `\n` | New line |
| `\t` | Tab |

The backslash introduces an escape sequence. `\n` moves the cursor to the next line; `\t` inserts a tab.

````python
# Original class notebook — cell 55
def calculation(a,b,**action_todo):
    answer = {}
    check_action = list(action_todo.keys())
    # print(check_action)
    if 'add' in check_action:
        answer['add'] = a+b
    if 'subtract' in check_action:
        answer['subtract'] = a-b
    if 'multiply' in check_action:
        answer['multiply'] = a*b
    if 'divide' in check_action:
        answer['divide'] = a/b
    return answer
````

````python
# Original class notebook — cell 56
calculation(10,20, add=True, subtract=True)
````

````python
# Original class notebook — cell 57
calculation(10,20, divide=True, subtract=True)
````
