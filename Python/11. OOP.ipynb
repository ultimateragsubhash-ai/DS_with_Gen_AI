# Part 1 — Classes and objects
<text>**Session source: 14 October 2026**</text>

## 1.1 What is a class?
A **class** is a blueprint that describes the data (attributes) and behaviour (methods) that objects created from it can have.

An **object** (also called an *instance*) is a concrete instance created from a class. Each object has its own identity and can hold its own instance attribute values.

```python
class Demo:
    pass

instance_1 = Demo()
instance_2 = Demo()

print(type(instance_1))
print(type(instance_2))
print(instance_1 is instance_2)
print(id(instance_1))
print(id(instance_2))
```

### Dry run
- `class Demo:` defines the class; it does not create `instance_1` or `instance_2`.
- `Demo()` creates a new instance.
- Each call to `Demo()` creates a distinct object.
- `type(instance_1)` reports the class of the object.
- `is` checks whether two references point to the same object.

## 1.2 A method is a function defined inside a class
A function defined inside a class is called a **method**. An object can call an instance method using dot notation.

```python
class Demo:
    def print_demo_detail(self):
        print("This is a method inside Demo.")

obj = Demo()
obj.print_demo_detail()
```

## 1.3 `self`: the current instance
`self` is the conventional name for the first parameter of an instance method. When you call `obj.method()`, Python automatically passes `obj` as that first argument.

`self` is not a reserved keyword; it is a strong Python convention. Using another name is technically possible, but avoid doing so.

```python
class Demo:
    def show_identity(self):
        print("Object id:", id(self))

obj_1 = Demo()
obj_2 = Demo()

obj_1.show_identity()
obj_2.show_identity()
print("obj_1 id:", id(obj_1))
print("obj_2 id:", id(obj_2))
```

The `self` received by `obj_1.show_identity()` refers to `obj_1`; in the second call it refers to `obj_2`. This is how one class method can operate on different objects.

## 1.4 Constructor / initializer: `__init__`
The class can initialize an object with values at creation time. In Python, `__init__` is the initializer method. It runs automatically after a new instance is created.

- `__init__` is not mandatory.
- Its name is special and must be written exactly.
- It runs for every new instance created.
- It is not normally called manually to “start” an already-created object again.

```python
class Demo:
    def __init__(self, a):
        print("Initializer received self id:", id(self))
        self.a = a

    def demo_1(self):
        print(self.a)

obj_1 = Demo(12)
obj_1.demo_1()

obj_2 = Demo(25)
obj_2.demo_1()
```

### Important distinction: defined once, executed per object
The method definition exists once in the class body. When objects are instantiated, `__init__` executes for each new object and initializes that object's attributes. Do not confuse “method defined once” with “initializer runs only once globally.”

## 1.5 Attributes and methods — smartphone example
An **attribute** stores information associated with an object. A **method** represents behaviour.

```python
class SmartPhone:
    def __init__(self, email, phone_no, name):
        self.email = email
        self.phone_no = phone_no
        self.name = name

    def open_chrome(self):
        print(f"Using {self.email} as login id")

    def open_facebook(self):
        print(f"Using {self.phone_no} as login id")

    def welcome_user(self):
        print(f"Hello User {self.name}")

obj_1 = SmartPhone("Monal@gmail.com", "+1232", "Monal")
obj_2 = SmartPhone("Rishabh@gmail.com", "+9132233", "Rishabh")

obj_1.open_chrome()
obj_1.open_facebook()
obj_1.welcome_user()

obj_2.open_chrome()
obj_2.open_facebook()
obj_2.welcome_user()
```

### What belongs where?
| Class concept | Smartphone example |
|---|---|
| Class | `SmartPhone` |
| Instance/object | `obj_1`, `obj_2` |
| Initializer | `__init__` |
| Attributes | `email`, `phone_no`, `name` |
| Methods | `open_chrome`, `open_facebook`, `welcome_user` |

The objects share the same class-defined behaviour, but each object stores its own attribute values.

## 1.6 Why OOP? Traditional approach versus object-based approach
The supplied game example first stores each player's name and health in separate variables, then passes health values to a function. As the number of players grows, keeping related values and operations coordinated becomes cumbersome.

```python
# Traditional style: data is stored separately.
char_1_name = "Sagar"
char_1_health = 100

char_2_name = "Likitha"
char_2_health = 90

def take_damage(health, amount):
    new_health = health - amount
    return new_health

char_1_health = take_damage(char_1_health, 20)
char_2_health = take_damage(char_2_health, 30)

print(f"{char_1_name} has {char_1_health} HP/Points left.")
print(f"{char_2_name} has {char_2_health} HP/Points left.")
```

With OOP, related data and the operations that act on that data are grouped together in a class.

```python
class Character:
    def __init__(self, name, health):
        self.name = name
        self.health = health

    def take_damage(self, amount):
        self.health -= amount

    def stats(self):
        print(f"{self.name} has {self.health} HP/Points left.")

player_1 = Character("Sagar", 100)
player_2 = Character("Likitha", 90)
player_3 = Character("Anushka", 90)

player_1.take_damage(20)
player_2.take_damage(30)
player_3.take_damage(30)

player_1.stats()
player_2.stats()
player_3.stats()

print(type(1))
print(type(player_1))
```

### Compare the two approaches
- Traditional: data is spread across variables; functions receive the data they need.
- OOP: each `Character` object carries its own `name` and `health`, and its methods operate on that object's state.
- OOP does not automatically make every program better. It is useful when the program has entities with related state and behaviour, and when organizing those responsibilities makes the code easier to extend.

# Part 3 — Inheritance
<text>**Session source: 17 October 2026**</text>

## 3.1 Meaning
Inheritance is about **sharing and extending code across multiple classes**. A child class (subclass) can reuse methods and attributes provided by a parent class (base class), and can add or override behaviour.

```python
class Animal:
    def __init__(self, legs):
        self.leg_count = legs

    def how_many_legs(self):
        print(f"Has {self.leg_count} legs.")

    def eat(self):
        print("I can eat!")

class Dog(Animal):
    def __init__(self, name, legs):
        self.name = name
        super().__init__(legs)

    def bark(self):
        print(f"{self.name} says woof!")

dog_1 = Dog("Ellie", 4)
dog_1.bark()
dog_1.how_many_legs()
dog_1.eat()
```

### Understand `super()`
`Dog` inherits from `Animal`. Its initializer needs to initialize both the child-specific `name` and the parent-provided `leg_count`.

`super().__init__(legs)` calls the next initializer in the method resolution order (MRO), which here is `Animal.__init__`.

```python
class Animal:
    def __init__(self, legs):
        self.leg_count = legs

    def eat(self):
        print("I can eat!")

class Cat(Animal):
    def __init__(self, name):
        self.name = name
        super().__init__(4)

    def meow(self):
        print(f"{self.name} says meow!")

cat_1 = Cat("Garfield")
cat_1.meow()
cat_1.eat()
print(cat_1.leg_count)
```

### Note about the supplied Cat example
The original notebook's `Cat.__init__` set `name` but did not call `Animal.__init__`. Therefore, `cat_1.eat()` worked because `eat()` did not need `leg_count`, but `cat_1.how_many_legs()` would fail because `leg_count` had never been initialized. The corrected example above calls `super().__init__(4)`.

## 3.2 Method overriding
A child class can provide its own implementation of a method inherited from the parent. This is called **overriding**.

```python
class Animal:
    def speak(self):
        return "Some generic animal sound"

class Dog(Animal):
    def speak(self):
        return "Woof"

class Cat(Animal):
    def speak(self):
        return "Meow"

for animal in [Dog(), Cat(), Animal()]:
    print(animal.speak())
```

## 3.3 Multiple inheritance
A class can inherit from more than one parent class. Python uses the Method Resolution Order (MRO) to determine the order in which it searches for methods.

```python
class Engine:
    def start(self):
        return "Starting engine"

class Electronics:
    def start(self):
        return "Starting electronics"

class Tesla(Engine, Electronics):
    pass

smart_car = Tesla()
print(smart_car.start())
print(Tesla.__mro__)
```

Because `Engine` appears before `Electronics` in `class Tesla(Engine, Electronics)`, the lookup finds `Engine.start` first. `Tesla.__mro__` displays the lookup order.

The supplied notebook also inspected `int.__mro__` to show that built-in Python types have inheritance hierarchies too.

```python
print(int.__mro__)
```

## 3.4 When inheritance is a poor fit
The class example uses `Bird` and `Penguin`. If every `Bird` inherits a `fly()` method, a penguin would incorrectly appear to have flying behaviour. This illustrates a design problem: inheritance should represent a valid “is-a” relationship.

```python
class Bird:
    def fly(self):
        print("Can fly!")

class Penguin(Bird):
    pass

penguin = Penguin()
penguin.fly()  # Technically works, but models the real-world behaviour incorrectly.
```

The issue is not Python syntax. The model itself is wrong: not every bird can fly. One option is to model shared capabilities separately and compose the capabilities an object actually has.

## 3.5 Composition
**Composition** means building an object using other objects as components. Instead of inheriting behaviour that may not apply, an object holds references to objects that provide the behaviour it needs.

```python
class Eat:
    def eat(self):
        print("Can eat!")

class Swim:
    def swim(self):
        print("Can swim!")

class Penguin(Eat):
    def __init__(self, swim_obj):
        self.swim_obj = swim_obj

    def swim(self):
        self.swim_obj.swim()

swim_obj = Swim()
penguin = Penguin(swim_obj)

penguin.eat()
penguin.swim()
```

### Composition example: Car, Engine and SoundSystem
The `Car` receives independent `Engine` and `SoundSystem` objects and delegates work to them. This is a “has-a” relationship: a car **has an** engine and **has a** sound system.

```python
class Engine:
    def __init__(self, horsepower):
        self.horsepower = horsepower

    def start(self):
        print(f"Engine with {self.horsepower} horsepower is started.")

class SoundSystem:
    def __init__(self, brand):
        self.brand = brand

    def play_music(self):
        print(f"Playing music through {self.brand} speakers.")

class Car:
    def __init__(self, company_name, model_name, engine_obj, sound_obj):
        self.company_name = company_name
        self.model_name = model_name
        self.engine = engine_obj
        self.sound_system = sound_obj

    def drive(self):
        print(f"Starting your {self.company_name} - {self.model_name}")
        self.engine.start()
        self.sound_system.play_music()

v8_engine = Engine(1000)
bose_sound_system = SoundSystem("Bose")

car_1 = Car("Ford", "Mustang GT", v8_engine, bose_sound_system)
car_1.drive()
```

### Inheritance versus composition
| Inheritance | Composition |
|---|---|
| Models an “is-a” relationship | Models a “has-a” relationship |
| Child reuses/extends parent behaviour | Object delegates to contained component objects |
| Can create tight coupling to parent design | Components can often be swapped independently |
| Useful when the subtype genuinely satisfies the parent contract | Useful when capabilities vary or should be assembled |

**Rule of thumb:** do not use inheritance merely to reuse a few lines of code. First check whether the child really is a valid kind of the parent.

# Part 4 — Encapsulation
<text>**Session source: 17 October 2026**</text>

## 4.1 Meaning
Encapsulation groups data and the methods that operate on that data inside a class. It also encourages callers to use the class's intended methods rather than changing internal state arbitrarily.

The session highlighted public, protected-style and private-style naming in Python.

```python
class Car:
    def __init__(self, brand):
        self.brand = brand                         # public
        self._ac_voltage = 30                      # non-public convention
        self.__current_gps_coordinate = "LA 1.23 LO 4.56"  # name-mangled

    def tell_brand(self):
        print(self.brand)

    def check_ac(self):
        return self._ac_voltage <= 30

    def run_ac(self):
        if self.check_ac():
            print("Starting AC")
        else:
            print("AC cannot start")

    def accessing_gps_coordinate(self, password):
        if password == "abc123":
            print(self.__current_gps_coordinate)
        else:
            print("Access denied")

    def update_coordinate(self, new_coordinate):
        self.__current_gps_coordinate = new_coordinate

car = Car("Toyota")
car.tell_brand()
print(car.brand)
print(car.check_ac())
car.accessing_gps_coordinate("wrong")
car.accessing_gps_coordinate("abc123")
car.update_coordinate("Updated location")
car.accessing_gps_coordinate("abc123")
```

## 4.2 Python naming conventions
| Form | Common meaning | What Python does |
|---|---|---|
| `name` | Public | Normal attribute access |
| `_name` | Internal / protected-style convention | Still accessible; underscore signals “please treat as non-public” |
| `__name` | Private-style attribute | Name mangling changes the stored attribute name |

A leading double underscore inside a class triggers **name mangling**. For example, `__current_gps_coordinate` in `Car` is stored under a name similar to `_Car__current_gps_coordinate`.

This is primarily to avoid accidental name clashes in subclasses, not to provide secure access control.

## 4.5 Name mangling helps avoid subclass name collisions
A parent and child can each define `__status`. Python mangles the names using the class name, so the two attributes do not overwrite one another.

```python
class Machine:
    def __init__(self):
        self.__status = "Machine ready"

    def get_machine_status(self):
        return self.__status

class Printer(Machine):
    def __init__(self):
        super().__init__()
        self.__status = "Printer ready"

    def get_printer_status(self):
        return self.__status

printer = Printer()
print(printer.get_machine_status())
print(printer.get_printer_status())

# Demonstration only: inspect the mangled attributes.
print(printer._Machine__status)
print(printer._Printer__status)
```

## 4.6 Getters, setters and validation
The class discussion mentions getter/setter methods as a way to expose controlled access. In Python, a `property` is often a clean way to provide attribute-like access while validating updates.

```python
class BankAccount:
    def __init__(self, person_id):
        self.person_id = person_id
        self._balance = 0

    @property
    def balance(self):
        return self._balance

    def deposit(self, amount):
        if amount <= 0:
            raise ValueError("Deposit amount must be positive")
        self._balance += amount

account = BankAccount(123)
account.deposit(100)
print(account.balance)
```

### Original BankAccount example — issue to notice
The 17 October notebook included `self.email_address = new_email.strip.lower`. That expression does not call either method and is missing parentheses. The corrected form is `new_email.strip().lower()`.

```python
class BankAccount:
    def __init__(self, person_id):
        self.person_id = person_id
        self.balance = 0
        self.email_address = ""

    def update_email_address(self, new_email):
        self.email_address = new_email.strip().lower()

ba_1 = BankAccount(123)
ba_1.update_email_address("  USER@EXAMPLE.COM  ")
print(ba_1.email_address)
```

### Encapsulation — practical interpretation
- Keep an object's state and the operations that maintain it together.
- Validate values at the point where state changes.
- Expose methods/properties that communicate supported operations.
- Do not assume `_name` or `__name` makes data secure against a determined caller.

# Part 5 — Polymorphism
<text>**Session source: 18 October 2026**</text>

## 5.1 Meaning
“Poly” means many and “morphism” means forms. The same operation can behave differently depending on the object or context.

A familiar Python example is `+`:
- With numbers, it performs addition.
- With strings, it performs concatenation.

```python
print(10 + 20)
print("Hello " + "Python")
```

In OOP, polymorphism supports **flexibility and uniform communication**: code can call the same method on different object types, while each object supplies its own implementation.

## 5.2 Payment gateway example — without polymorphism
The first approach uses conditional branches for each payment gateway. As new gateways are added, the checkout function needs more branches.

```python
def checkout(gateway, amount):
    if gateway == "paypal":
        print(f"Charging {amount} via PayPal")
    elif gateway == "gpay":
        print(f"Charging {amount} via GPay")
    elif gateway == "credit_card":
        print(f"Charging {amount} via credit card")
    else:
        print(f"Unsupported gateway: {gateway}")

checkout("gpay", 200)
checkout("paypal", 200)
checkout("crypto", 200)
```

## 5.3 Same method name across different classes
Each payment class implements `process_payment`. The checkout function does not need to know the internal payment steps; it only calls the shared method.

```python
class PayPal:
    def process_payment(self, amount):
        return f"Charging {amount} securely via PayPal"

class GPay:
    def process_payment(self, amount):
        return f"Charging {amount} securely via GPay"

class CreditCard:
    def process_payment(self, amount):
        return f"Charging {amount} securely via CreditCard"

def execute_checkout(payment_gateway, amount):
    result = payment_gateway.process_payment(amount)
    print(result)

paypal = PayPal()
gpay = GPay()
credit_card = CreditCard()

execute_checkout(paypal, 200)
execute_checkout(gpay, 200)
execute_checkout(credit_card, 200)
```

### Key idea
`execute_checkout()` is written once. It works with any object that provides a compatible `process_payment(amount)` method. This is an example of **duck typing**: the relevant question is whether the object supports the required operation, not whether it belongs to a particular named class.

## 5.4 Registry / dictionary approach from the class
A dictionary can map user-facing gateway names to gateway objects. The function looks up the object and invokes the common method.

```python
payment_registry = {
    "paypal": PayPal(),
    "gpay": GPay(),
    "credit_card": CreditCard(),
}

def process_payment(gateway_name, amount):
    gateway = payment_registry.get(gateway_name)

    if gateway is None:
        return f"Payment gateway {gateway_name} is not supported."

    return gateway.process_payment(amount)

print(process_payment("gpay", 200))
print(process_payment("paypal", 350))
print(process_payment("unknown", 100))
```

### Improvement over a broad `try/except`
The supplied script used `.get(...).process_payment(...)` inside a bare `except`. If the key is absent, `.get()` returns `None` and calling a method on it fails. A direct `None` check is clearer and avoids hiding unrelated programming errors.

The supplied script also encoded an error marker (`-1`) inside a comma-separated message and split the string to detect failure. Returning a clear value or raising a specific exception is easier to maintain.

## 5.5 Type hints
Type hints document the expected types of parameters and return values. By default, Python does not enforce annotations at runtime.

```python
def add(value_1: float, value_2: float) -> float:
    return value_1 + value_2

print(add(1, 2))
print(add(1.5, 2.5))
```

`value_1: float` and `value_2: float` are annotations. They help readers, IDEs and static type checkers. They do not automatically prevent a caller from passing a string at runtime.

## 5.6 `Protocol`: describe the shape an object should provide
The supplied junior-developer example uses `typing.Protocol`. A protocol describes the methods an object is expected to provide. It supports structural typing: a class can satisfy the protocol by having the required methods; it does not have to inherit from the protocol explicitly.

```python
from typing import Protocol

class PaymentProcessor(Protocol):
    def process_payment(self, amount: float) -> str:
        ...

class PayPal:
    def process_payment(self, amount: float) -> str:
        return f"Charging {amount} securely via PayPal"

class GPay:
    def process_payment(self, amount: float) -> str:
        return f"Charging {amount} securely via GPay"

class CreditCard:
    def process_payment(self, amount: float) -> str:
        return f"Charging {amount} securely via CreditCard"

def process_with_gateway(payment_gateway: PaymentProcessor, amount: float) -> str:
    return payment_gateway.process_payment(amount)

print(process_with_gateway(PayPal(), 200))
print(process_with_gateway(GPay(), 200))
print(process_with_gateway(CreditCard(), 200))
```

### A mismatch in the supplied junior-developer script
The supplied `DebitCard` class defined `process()` rather than `process_payment()`. It therefore did not satisfy the expected interface and would fail when passed to the checkout function. The corrected version uses the same method name.

```python
class DebitCard:
    def process_payment(self, amount: float) -> str:
        return f"Charging {amount} securely via DebitCard"

print(process_with_gateway(DebitCard(), 200))
```

### `Protocol` versus runtime enforcement
- A `Protocol` primarily provides a contract for static type checkers and IDEs.
- It does not automatically validate every object at runtime.
- The key requirement here is that the object provides a callable `process_payment` method with a compatible signature.

# Part 6 — Abstraction
<text>**Session source: 18 October 2026**</text>

## 6.1 Meaning
Abstraction focuses on **what an object must provide**, rather than exposing every detail of how it does the work. It helps establish a common interface and standard across implementations.

The class explained that abstraction is not required to make a tiny piece of code run; it is a design practice that helps teams create consistent, maintainable systems.

## 6.2 Abstract Base Classes (ABC)
Python's `abc` module provides `ABC` and `@abstractmethod`.
- Inherit from `ABC` to create an abstract base class.
- Mark required methods with `@abstractmethod`.
- A concrete child class must implement all inherited abstract methods before it can be instantiated.

```python
from abc import ABC, abstractmethod

class NotificationService(ABC):
    @abstractmethod
    def send_message(self, message):
        pass

    @abstractmethod
    def read_message(self, message):
        pass

    def add(self):
        print("This is a regular method on the base class.")
```

`send_message` and `read_message` are required methods. `add` is a normal concrete method in this example; because it is not decorated with `@abstractmethod`, child classes are not required to override it.

## 6.3 Concrete notification implementations
Each notification class follows the same required interface, but its implementation differs.

```python
class EmailNotification(NotificationService):
    def send_message(self, message):
        return f"Sending email with message: {message}"

    def read_message(self, message):
        return f"Reading email message: {message}"

class SMSNotification(NotificationService):
    def send_message(self, message):
        return f"Sending SMS with message: {message}"

    def read_message(self, message):
        return f"Reading SMS message: {message}"

class WhatsappNotification(NotificationService):
    def send_message(self, message):
        return f"Sending WhatsApp message: {message}"

    def read_message(self, message):
        return f"Reading WhatsApp message: {message}"

class SlackNotification(NotificationService):
    def send_message(self, message):
        return f"Sending Slack message: {message}"

    def read_message(self, message):
        return f"Reading Slack message: {message}"
```

```python
services = [
    EmailNotification(),
    SMSNotification(),
    WhatsappNotification(),
    SlackNotification(),
]

for service in services:
    print(service.send_message("Your order has been shipped"))
    print(service.read_message("Your order has been shipped"))
```

## 6.4 What if a child does not implement every abstract method?
The class cannot be instantiated while abstract methods remain unimplemented. This is how an abstract base class enforces the blueprint.

```python
class IncompleteNotification(NotificationService):
    def send_message(self, message):
        return "Sending message"

# Expected TypeError if run:
# incomplete = IncompleteNotification()
```

## 6.6 `Protocol` versus `ABC`
| `Protocol` | Abstract Base Class (`ABC`) |
|---|---|
| Describes the required shape/operations | Defines a base class with abstract methods |
| Structural typing: compatible methods can be enough | Nominal inheritance is the usual usage |
| Primarily helps static type checking | Prevents instantiation until abstract methods are implemented |
| Does not normally force inheritance | Child class explicitly inherits from the ABC |

Both can describe interfaces, but they serve somewhat different purposes.

# Part 7 — Putting the four pillars together
A payment processing design can use:
- **Abstraction** to define the required `process_payment` interface.
- **Polymorphism** to let checkout call that method on different payment implementations.
- **Encapsulation** to keep provider-specific state and operations inside each class.
- **Composition** to let a checkout service hold or receive a payment processor object.

Inheritance is used when one class is genuinely a subtype of another; composition is often a better way to plug interchangeable components into a service.

```python
from abc import ABC, abstractmethod

class PaymentMethod(ABC):
    @abstractmethod
    def process_payment(self, amount):
        pass

class CardPayment(PaymentMethod):
    def __init__(self, last_four_digits):
        self._last_four_digits = last_four_digits

    def process_payment(self, amount):
        return f"Processed {amount} using card ending {self._last_four_digits}"

class WalletPayment(PaymentMethod):
    def __init__(self, wallet_name):
        self._wallet_name = wallet_name

    def process_payment(self, amount):
        return f"Processed {amount} using {self._wallet_name}"

class CheckoutService:
    def __init__(self, payment_method):
        self.payment_method = payment_method

    def checkout(self, amount):
        return self.payment_method.process_payment(amount)

checkout_1 = CheckoutService(CardPayment("1234"))
checkout_2 = CheckoutService(WalletPayment("GPay"))

print(checkout_1.checkout(500))
print(checkout_2.checkout(750))
```

## 8.1 Suggested design before coding
A reasonable design is:
- `Student`: represents one student and its details.
- `StudentRepository` (or `StudentManager`): stores and queries multiple student objects.
- `StudentExporter` / export method: writes selected records to CSV.
- A menu loop: handles user interaction and delegates to manager methods.

This is a suggested decomposition to help solve the assignment; the brief does not prescribe exact class names or an exact internal architecture.

```python
import csv

class Student:
    def __init__(self, name, standard, phone, address):
        self.name = name.strip()
        self.standard = standard.strip()
        self.phone = phone.strip()
        self.address = address.strip()

class StudentManager:
    def __init__(self):
        self._students = []

    def add_student(self, student):
        if not isinstance(student, Student):
            raise TypeError("student must be a Student object")
        self._students.append(student)

    def get_names_by_class(self, standard):
        return [
            student.name
            for student in self._students
            if student.standard.casefold() == standard.strip().casefold()
        ]

    def get_name_and_phone_by_class(self, standard):
        return [
            (student.name, student.phone)
            for student in self._students
            if student.standard.casefold() == standard.strip().casefold()
        ]

    def export_data(self, standard, filename=None):
        safe_standard = standard.strip().replace(" ", "_")
        if filename is None:
            filename = f"{safe_standard}_students.csv"

        matching_students = [
            student
            for student in self._students
            if student.standard.casefold() == standard.strip().casefold()
        ]

        with open(filename, "w", newline="", encoding="utf-8") as csv_file:
            writer = csv.writer(csv_file)
            writer.writerow(["name", "standard", "phone", "address"])
            for student in matching_students:
                writer.writerow([
                    student.name,
                    student.standard,
                    student.phone,
                    student.address,
                ])

        return filename
```

### Where could abstraction fit into the assignment?
If the assignment expects an abstract blueprint, you could define an abstract repository or exporter interface, then implement it in a concrete class. Do not add abstraction merely for decoration: use it where there is a real contract that multiple implementations must follow.

```python
from abc import ABC, abstractmethod

class StudentDataExporter(ABC):
    @abstractmethod
    def export_data(self, students, filename):
        pass

class CSVStudentExporter(StudentDataExporter):
    def export_data(self, students, filename):
        with open(filename, "w", newline="", encoding="utf-8") as csv_file:
            writer = csv.writer(csv_file)
            writer.writerow(["name", "standard", "phone", "address"])
            for student in students:
                writer.writerow([student.name, student.standard, student.phone, student.address])
        return filename
```

```python
class Book:
    def __init__(self, title, author, price):
        self.title = title
        self.author = author
        self.price = price

    def display(self):
        print(f"{self.title} by {self.author} — ₹{self.price}")

book = Book("Python Basics", "A. Writer", 450)
book.display()


class SafeBankAccount:
    def __init__(self, opening_balance=0):
        if opening_balance < 0:
            raise ValueError("Opening balance cannot be negative")
        self._balance = opening_balance

    @property
    def balance(self):
        return self._balance

    def deposit(self, amount):
        if amount <= 0:
            raise ValueError("Deposit must be positive")
        self._balance += amount

    def withdraw(self, amount):
        if amount <= 0:
            raise ValueError("Withdrawal must be positive")
        if amount > self._balance:
            raise ValueError("Insufficient balance")
        self._balance -= amount

account = SafeBankAccount(1000)
account.deposit(250)
account.withdraw(300)
print(account.balance)
```
