## 2. What is a `while` loop?

A `while` loop repeatedly executes a block of code **while a condition remains `True`**.

The condition is checked before each iteration:
- If it is `True`, Python enters the loop body.
- If it is `False`, Python skips the body and continues after the loop.
- After the body finishes, Python checks the condition again.

Unlike a `for` loop over a fixed range, a `while` loop does not inherently tell you how many times it will execute. You control when it stops by changing the condition or the values used by the condition.

General syntax:

```python
while condition:
    # statements to repeat
```

````python
start = 0

while start < 5:
    print("Hello", start)
    start += 1

print("Loop finished")
````

### 2.1 Dry run: `while start < 5`

| Check | `start` before body | Is `start < 5` true? | Action |
|---:|---:|---|---|
| 1 | 0 | True | Print 0; increase to 1 |
| 2 | 1 | True | Print 1; increase to 2 |
| 3 | 2 | True | Print 2; increase to 3 |
| 4 | 3 | True | Print 3; increase to 4 |
| 5 | 4 | True | Print 4; increase to 5 |
| 6 | 5 | False | Exit the loop |

The increment `start += 1` is important. It moves the program towards a point where the condition becomes false.

````python
# Same pattern as the class notebook, using 10 iterations
start = 0
print("While loop : start")

while start < 10:
    print("Hello", start)
    start += 1

print("While loop : end")
````

## 3. Conditions and Boolean values

The condition in a `while` loop produces a Boolean result: `True` or `False`.

The session connects this to concepts already learned:
- `if` / `else` statements
- Comparison operators such as `>`, `<` and `==`
- Boolean values `True` and `False`

A condition that uses variables requires those variables to exist before Python evaluates the condition.

````python
# Variables must be initialized before they are used in a condition.
a = 10
b = 5

if a > b:
    print(True)

# If a or b has not been defined, evaluating a > b raises NameError.
````

### 3.1 Truthy and falsy values

Python can also evaluate values directly in a condition. The session demonstrates that:
- `None` is falsy.
- `0` is falsy.
- A non-zero number such as `1` or `-11000` is truthy.
- `True` is truthy.
- `False` is falsy.

For the examples below, the `if` body runs only when the condition is truthy.

````python
if None:
    print("This will not print")

if -11000:
    print("A non-zero number is truthy")

if 1:
    print("1 is truthy")

if 0:
    print("This will not print")

if True:
    print("True is truthy")

if False:
    print("This will not print")
````

**Useful rule:** for numeric values, zero is falsy and non-zero values are truthy. `None` and `False` are also falsy. Empty strings and empty collections are falsy as well, although those examples were not the main focus of this class.

Do not confuse truthiness with a comparison. `if 2:` does not ask whether 2 is greater than or equal to something; it simply checks whether the value is truthy.

## 4. Infinite loops and how to avoid them

A loop becomes infinite when its condition remains true and nothing in the loop changes the state needed to make it false.

For example, if `start` remains `0`, then `start < 10` will always be true. The loop will keep printing and consuming resources.

The class explicitly warns that an infinite loop can consume CPU and RAM. Be careful when testing one.

````python
# Do NOT run this cell as written: it is an infinite loop.
# start = 0
# while start < 10:
#     print("Hello", start)
#
# start never changes, so start < 10 remains True forever.
````

### 4.1 Correcting the loop

Add an update inside the loop body. In this example, `start += 1` gradually changes the value until the condition becomes false.

````python
start = 0

while start < 10:
    print("Hello", start)
    start += 1

print("Stopped because start is now 10")
````

### 4.2 A `while True` loop

The class also discusses writing `while True`. This is an intentional loop that keeps running until the program changes its flow or exits the loop through some other mechanism.

The notebook's example uses a Boolean variable (`stop`) instead of a literal `True`. This gives the program a variable that can be changed to `False` when the user wants to stop.

````python
# Safe demonstration of a Boolean-controlled while loop
stop = True
counter = 0

while stop:
    print("Running:", counter)
    counter += 1

    if counter == 3:
        stop = False

print("Stopped")
````

### 4.3 Why initialize the control variable first?

The session explains this design:
1. Create a Boolean variable and set it to `True`.
2. Use that variable as the `while` condition.
3. Inside the loop, write a condition that can change the variable to `False`.
4. When it becomes false, the next condition check exits the loop.

If the variable is false before the loop starts, the body will not execute even once.

````python
stop = False

while stop:
    print("This body will not run")

print("The loop was skipped because stop was False")
````

## 5. A user-controlled addition calculator

The instructor builds a calculator that keeps accepting numbers and adds them to a running total. The user can choose when to stop.

The intended behavior:
1. Start with a total of `0`.
2. Ask the user for a number.
3. Convert the input to `float` so decimal numbers are accepted too.
4. Add the entered number to the running total.
5. Ask whether the user wants to continue.
6. Normalize the answer with `.lower()` so both `Y` and `y`, or `N` and `n`, are handled consistently.
7. If the answer is `n`, change the control variable to `False`.

### 5.1 The class notebook's simpler accumulation example

This version keeps asking for numbers and adding them. It stops when the user enters `n`.

**Run this cell interactively:** it waits for keyboard input.

````python
start = 0
stop = True

while stop:
    val = float(input("Enter value you want to add: "))
    start += val
    print("Current total:", start)

    user_continue = input(
        "Do you wish to continue? y for Yes, n for No: "
    ).lower()

    if user_continue == "n":
        stop = False

print("Final total:", start)
````

| Next condition check | `while stop` is false | 400 | False |

The loop exits and the final total is 400.

### 5.6 Calculator assignment from the class

The instructor's assignment is to build a calculator that:
- Runs inside a loop.
- Takes a number from the user.
- Asks which operation to perform:
  - `1` for addition
  - `2` for multiplication
  - `3` for subtraction
  - `4` for division
- Performs the selected operation.
- Asks whether the user wants to continue.

The class notebook contains this as an exercise prompt, not a completed implementation. A sample practice implementation is provided below so you can compare your own attempt.

````python
# Practice implementation of the class calculator assignment
result = float(input("Enter starting number: "))
stop = True

while stop:
    operation = input(
        "Choose operation: 1-add, 2-multiply, 3-subtract, 4-divide: "
    )
    value = float(input("Enter next number: "))

    if operation == "1":
        result += value
    elif operation == "2":
        result *= value
    elif operation == "3":
        result -= value
    elif operation == "4":
        if value == 0:
            print("Cannot divide by zero")
        else:
            result /= value
    else:
        print("Invalid operation")

    print("Current result:", result)

    choice = input("Continue? (Y/N): ").lower()
    if choice == "n":
        stop = False

print("Final result:", result)
````

## 8. Why functions are useful

The instructor compares functions with reusable features in software: instead of writing the same implementation everywhere, a developer can define the behavior once and call it where needed.

The class first shows what repeated code looks like without a function. The same three statements are copied for different pairs of values.

````python
# Repeated code — first set of values
a = 10
b = 22
print("function start")
print("addition of 2 values is:", a + b)
print("add and print is done")

# Repeated code — second set of values
a = 100
b = 22
print("function start")
print("addition of 2 values is:", a + b)
print("add and print is done")
````

The same behavior can be placed inside a function and called with different values. This reduces repetition and makes the code reusable.

## 9. Function inputs: parameters and arguments

The instructor uses a memorable analogy: the function has “hands” to receive values.

The technical terms are:
- **Parameter:** the variable named in the function definition, inside the parentheses.
- **Argument:** the actual value supplied when the function is called.

For example, in:

```python
def add(a, b):
    return a + b

add(10, 20)
```

- `a` and `b` are parameters.
- `10` and `20` are arguments.

The arguments are assigned to the parameters in positional order: `a = 10` and `b = 20`.

````python
def add(a, b):
    print("a =", a)
    print("b =", b)
    print("sum =", a + b)

add(10, 1)
````

### 9.1 A function with no parameters cannot receive positional arguments

A function defined with empty parentheses has no parameters. Calling it with two positional arguments raises `TypeError`.

````python
def say_hello_5():
    for i in range(5):
        print("hello")

say_hello_5()

# This is not valid for this function:
# say_hello_5(10, 20)
````

````python
def add_v2(a, b):
    print(a + b)
    print("addition")

add_v2(10, 1)
````

## 10. Printing versus returning

This is one of the most important distinctions in the session.

**`print()` displays something.** It does not send that displayed value back to the caller as the function result.

**`return` sends a value back to the place where the function was called.** The caller can store it in a variable, print it, or use it in another calculation.

If a function finishes without an explicit `return`, Python returns `None` automatically.

### 10.1 Function prints a result but does not return it

The function below prints the sum. Since it has no explicit `return`, its returned value is `None`.

The notebook displays the `None` value because it is captured in `value` and then printed.

````python
def add_v2(a, b):
    print(a + b)
    print("addition")

value = add_v2(100, 1)
print("Returned value:", value)
print(type(value))
````

### 10.2 Function returns a result

The class's `add_v3(a, b)`:
1. Accepts two parameters.
2. Creates an internal variable `c = 100`.
3. Prints `a + b + c`.
4. Prints the word `"addition"`.
5. Returns only `a + b`.

For the call `add_v3(10, 20)`, it prints `130` and returns `30`. These are two different results because the printed calculation includes `c`, while the return expression does not.

````python
def add_v3(a, b):
    c = 100
    print(a + b + c)
    print("addition")
    return a + b

value = add_v3(10, 20)
print("Returned value:", value)
````

### 10.3 Dry run of `add_v3(10, 20)`

| Step | What happens | Value |
|---:|---|---:|
| 1 | Call `add_v3(10, 20)` | `a = 10`, `b = 20` |
| 2 | Set `c = 100` | `c = 100` |
| 3 | Evaluate `a + b + c` and print it | `130` |
| 4 | Print `"addition"` | `addition` |
| 5 | Evaluate `return a + b` | `30` |
| 6 | Assign returned value to `value` | `value = 30` |
| 7 | Print `value` | `30` |

The returned value is what the caller receives. The earlier `print()` statements are output generated while the function is running.

### 10.4 Capture a returned value before printing it

A returned value can be stored in a variable and then used later. If you call a function that returns something but do not print or otherwise use that result, the value is still returned to the caller, but you may not see it in a normal Python script.

Jupyter Notebook displays the value of the final expression in a cell automatically. A normal `.py` script does not automatically display a returned value just because a function call is the last line.

````python
def add_v3(a, b):
    return a + b

result = add_v3(10, 20)
print(result)
print(result * 2)
````

### 10.5 A function with no parameters but a return value

The class's `add_v4()` example returns the string `"hello"` without accepting any parameters.

This is possible because receiving input and returning output are independent choices. A function can:
- Have no parameters and no explicit return.
- Have parameters and no explicit return.
- Have parameters and return a value.
- Have no parameters and return a value.

## 12. Functions can accept and return collections

The class demonstrates a function that accepts a list, loops over its values, prints them, and returns a list containing three values.

The parameter `list_a` refers to the list passed when the function is called. The function returns `["done", "done", 123]`, and the caller stores that returned list in `value_4`.

````python
def to_loop_over_values(list_a):
    for i in list_a:
        print(i)

    return ["done", "done", 123]

value_4 = to_loop_over_values([1, 2, 3, 4, 4, 5667])

print(value_4)
print(type(value_4))
````

### 12.1 Dry run of `to_loop_over_values(...)`

1. The function is called with `[1, 2, 3, 4, 4, 5667]`.
2. The list is assigned to the parameter `list_a`.
3. The `for` loop prints each item in that list.
4. After the loop finishes, the function evaluates its `return` statement.
5. The returned list `["done", "done", 123]` is assigned to `value_4`.
6. The caller prints the returned list and its type.

The returned list is not the same as the input list. The function returns the exact list written in its `return` statement.

## 13. Returning values makes functions reusable in expressions

The last notebook example defines a simple `add(a, b)` function that returns a sum. Since it returns a number, the function calls can be used inside a larger arithmetic expression.

````python
def add(a, b):
    return a + b

answer = add(1, 2) + add(3, 4)
print(answer)
````

### 13.1 Dry run
