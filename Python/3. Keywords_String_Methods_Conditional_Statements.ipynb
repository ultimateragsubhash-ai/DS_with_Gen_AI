### 1.1 Two ways to write Python code

| File type | Extension | How code is executed |

### 1.2 Why use a notebook?

A notebook is useful while learning, experimenting, and testing individual pieces of code.

- You can run one cell without running every other cell.
- If one cell produces an error, other independent cells can still be run.
- A later cell may still fail if it depends on a variable or result that was not created correctly in an earlier cell.

A `.py` script is commonly used when the program is ready to run as a complete program.

````python
# Example: separate notebook cells can be run independently.

first_value = 10
print(first_value)
````

````python
# This cell is independent of the previous cell.
message = "Testing one cell at a time"
print(message)
````

**Important:** Notebook cells share a kernel/session. Variables created in one cell remain available to later cells until the kernel is restarted or the variable is changed. Therefore, execution order can matter even though cells can be run separately.

## 3. Python keywords

### 3.1 What is a keyword?

As explained in class, Python keywords are **special words that are part of the Python language itself**.

Keywords have predefined meaning in Python syntax. They are not ordinary names that you can freely use as variable names.

Examples seen in the class discussion include:
- `if`
- `else`
- `and`
- `True`
- `False`
- `None`
- `import`
- `as`

### 3.2 Display Python's keyword list

The original notebook uses Python's built-in `keyword` module:

````python
import keyword

print(keyword.kwlist)
````

`keyword.kwlist` provides the list of keywords recognized by the Python interpreter running this notebook.

The instructor used an analogy: `import keyword` is like bringing a book from a library, and `keyword.kwlist` extracts the list of keywords from that book.

### 3.3 Check whether a word is a keyword

The same module provides `iskeyword()`:

````python
import keyword

print(keyword.iskeyword("if"))
print(keyword.iskeyword("keyword"))
print(keyword.iskeyword("import"))
````

The word `keyword` is the module name in this example; it is not itself a Python keyword. `if` and `import` are keywords.

### 3.4 Why keywords matter

- They are part of the language syntax.
- Their meanings are predefined.
- You should not use a keyword as a variable name.
- Python is case-sensitive: `True` is a keyword, while `true` is a different name.

````python
# Try these separately:
print(keyword.iskeyword("True"))
print(keyword.iskeyword("true"))
````

## 4. String methods and whitespace

The class described a string as a Python data type and introduced methods (the instructor initially called them an object's “abilities”).

A method is accessed using dot notation:

```python
some_string.method_name()
```

The original notebook uses this string:

````python
kn_academy_batch_name = "   Ultimate Data Science & AI Engineer Bootcamp 3.0 "
print(kn_academy_batch_name)
````

Notice that the string contains spaces before and after the visible text.

### 4.1 `strip()` — remove spaces from both ends

The `strip()` method removes leading and trailing whitespace from a string.

````python
kn_academy_batch_name = "   Ultimate Data Science & AI Engineer Bootcamp 3.0 "

clean_name = kn_academy_batch_name.strip()
print(clean_name)
````

Expected visible result:

```text
Ultimate Data Science & AI Engineer Bootcamp 3.0
```

`strip()` removes whitespace at the beginning and at the end. It does not remove spaces between words.

### 4.2 `rstrip()` — remove spaces from the right

The class introduced `rstrip()` as the method that removes whitespace from the right-hand end of a string.

````python
course_name = "   Ultimate Data Science & AI Engineer Bootcamp 3.0 "
right_cleaned = course_name.rstrip()

print(repr(right_cleaned))
````

`repr()` is used here so that any remaining leading spaces are easier to see.

### 4.3 `lstrip()` — remove spaces from the left

The class introduced `lstrip()` as the method that removes whitespace from the left-hand end.

````python
course_name = "   Ultimate Data Science & AI Engineer Bootcamp 3.0 "
left_cleaned = course_name.lstrip()

print(repr(left_cleaned))
````

### 4.4 Use `lstrip()` and `rstrip()` instead of `strip()`

One class exercise asked learners to remove spaces from both ends without using `strip()`.

**Technique 1 — assign the result after each operation**

````python
kn_academy_batch_name = "   Ultimate Data Science & AI Engineer Bootcamp 3.0 "

kn_academy_batch_name = kn_academy_batch_name.lstrip()
kn_academy_batch_name = kn_academy_batch_name.rstrip()

print(kn_academy_batch_name)
````

**Technique 2 — chain the methods**

````python
kn_academy_batch_name = "   Ultimate Data Science & AI Engineer Bootcamp 3.0 "

kn_academy_batch_name = kn_academy_batch_name.lstrip().rstrip()
print(kn_academy_batch_name)
````

Both techniques produce a string with the leading and trailing spaces removed.

### 4.5 Why these two lines alone do not update the variable

The original notebook intentionally included this example:

````python
kn_academy_batch_name = "   Ultimate Data Science & AI Engineer Bootcamp 3.0 "

kn_academy_batch_name.lstrip()
kn_academy_batch_name.rstrip()

# The variable itself has not been reassigned.
print(repr(kn_academy_batch_name))
````

String methods return a result; they do not modify the original string in place. Strings are immutable. To keep the cleaned result, assign the returned value back to a variable.

### 4.6 Methods depend on the data type

The original notebook included an integer example:

````python
a = 12

# This raises AttributeError because an integer does not have strip():
# a.strip()
````

The instructor's point was that methods available on one data type are not automatically available on another data type. `strip()` is a string method, so it works on strings, not integers.

### 4.7 Python is case-sensitive

Method names must use the correct capitalization:

````python
sample_text = "  Python  "

print(sample_text.strip())

# Incorrect capitalization — uncomment to see the error:
# sample_text.STRIP()
````

## 5. Control flow and conditional statements

The instructor described **control flow** as determining the order in which code executes.

A conditional statement allows a program to choose what to execute based on whether a condition is `True` or `False`.

### 5.1 Basic `if` syntax

The syntax shown in class:

````python
# General syntax (template):
# if condition:
#     statement_to_execute
````

How to read it:

1. Python evaluates the condition.
2. If the condition is `True`, Python enters the indented block and executes its statements.
3. If the condition is `False`, Python skips that block and continues after it.

### 5.2 Indentation defines the block

The instructor emphasized that statements indented under an `if` belong to that `if` block.

The class discussed using the Tab key for indentation. The exact visual width can depend on the editor settings; the important Python rule is consistent indentation within a block.

````python
a = 3

if a > 3:
    print(8)

print("Hello world")
````

**Dry run**

- `a` is `3`.
- `a > 3` evaluates to `False`.
- Python skips `print(8)`.
- Python continues to the next unindented statement.
- Output: `Hello world`.

### 5.3 A true condition with multiple statements in the block

The transcript demonstrates the same idea with a value of `a = 5` and several statements inside the `if` block:

````python
a = 5

if a > 3:
    print(a)
    a = a + 1
    print(a)

print("Hello world")
````

**Dry run**

| Step | What happens | Value of `a` |
|---:|---|---:|
| 1 | `a = 5` | 5 |
| 2 | Check `a > 3` → `5 > 3` is `True` | 5 |
| 3 | `print(a)` | Prints 5 |
| 4 | `a = a + 1` | 6 |
| 5 | `print(a)` | Prints 6 |
| 6 | Leave the indented block and continue | 6 |
| 7 | `print("Hello world")` | Prints Hello world |

All three indented statements are part of the `if` block.

## 6. `if` and `if-else` statements

### 6.1 Age eligibility example

The original notebook contains this example:

````python
# Code that decides whether age is suitable for a party
age = float(input("Enter your age: "))

if age >= 18:
    print("You can go for a party")
else:
    print("You cannot go for a party")

print("Decision made")
````

How it works:

- If `age >= 18` is `True`, the first message is printed.
- Otherwise, the `else` block is executed.
- `print("Decision made")` is outside both blocks, so it runs in either case.

### 6.2 General `if-else` structure

````python
# General syntax (template):
# if condition:
#     statements_when_true
# else:
#     statements_when_false
````

`else` does not have a condition of its own. It handles the case where the preceding `if` condition is false.

### 6.3 Important distinction: statements inside and outside the block

````python
age = 17

if age >= 18:
    print("Eligible")

print("This line runs whether the condition is True or False")
````

The final `print()` is not indented, so it is outside the `if` block and runs regardless of the condition.

## 7. `if-elif-else`: student marks categorization

The class problem statement was to categorize students based on marks.

The ranges shown in the class notes and original notebook are:

| Marks | Category/message |
|---:|---|
| 85 to 100 | Brilliant |
| 75 to below 85 | Great |
| 60 to below 75 | Good |
| Other values | Please do better next time |

The notebook code uses `float(input(...))` to receive marks.

````python
marks = float(input("Enter student marks: "))

if marks >= 85 and marks <= 100:
    print("Brilliant")
elif marks >= 75 and marks < 85:
    print("great")
elif marks >= 60 and marks < 75:
    print("Good")
else:
    print("Please do better next time.")
````

### 7.1 Why use `elif`?

The instructor first discussed why several independent `if` statements are not the same as a single connected decision chain.

- `if` checks the first condition.
- `elif` checks another condition only when the earlier condition(s) were false.
- `else` handles the remaining case when none of the earlier conditions matched.

In a connected `if-elif-else` chain, once one branch is selected, the remaining branches in that chain are skipped.

### 7.2 Dry-run examples

Use the class's marks ranges to trace these values:

| Input marks | First condition | Second condition | Third condition | Result |
|---:|---|---|---|---|
| 93 | True | Not checked | Not checked | Brilliant |
| 76 | False | True | Not checked | great |
| 65 | False | False | True | Good |
| 50 | False | False | False | Please do better next time |

**Range note:** The source notebook's conditions do not explicitly validate a lower bound of 0. Negative marks therefore reach the `else` branch, as do marks below 60 and values above 100.

## 8. Nested `if`

The class then introduced a nested conditional: an `if` statement placed inside another `if` or `elif` block.

The original notebook uses this version of the marks example:

````python
marks = float(input("Enter student marks: "))

if marks >= 85 and marks <= 100:
    if marks > 90:
        print("Brilliant ++")
    else:
        print("Brilliant")
elif marks >= 75 and marks < 85:
    print("great")
elif marks >= 60 and marks < 75:
    print("Good")
else:
    print("Please do better next time.")
````

### 8.1 How the nested condition works

1. Python first checks whether marks are between 85 and 100.
2. Only if that outer condition is true does Python enter its indented block.
3. Inside that block, Python checks whether marks are greater than 90.
4. If `marks > 90`, it prints `Brilliant ++`.
5. Otherwise, it prints `Brilliant`.
6. If the outer condition is false, Python continues to the `elif` branches.

### 8.2 Dry-run examples

| Marks | Outer condition (`85–100`) | Inner condition (`> 90`) | Output |
|---:|---|---|---|
| 93 | True | True | Brilliant ++ |
| 88 | True | False | Brilliant |
| 76 | False | Not checked | great |
| 65 | False | Not checked | Good |
| 50 | False | Not checked | Please do better next time |

A nested block is simply a block inside another block. The indentation shows which statements belong to which condition.

## 10. Quick revision

### Keywords
- Keywords are special words that are part of the Python language.
- `import keyword` imports the keyword module.
- `keyword.kwlist` displays the interpreter's keyword list.
- `keyword.iskeyword("if")` checks whether a word is a keyword.

### Notebook and script
- `.py` is a Python script.
- `.ipynb` is an interactive notebook.
- Notebook cells can be executed separately.
- Cells can still depend on variables created by earlier cells.
- `ipykernel` enables a Python environment to run notebook cells.

### String methods
- `strip()` removes whitespace from both ends.
- `lstrip()` removes whitespace from the left.
- `rstrip()` removes whitespace from the right.
- String methods return a result; assign it if you want to keep the cleaned value.
- `strip()` is a string method; it is not available on integers.
- Python is case-sensitive.

### Conditional statements
- `if` executes its block when the condition is `True`.
- `else` handles the false case.
- `elif` checks another condition in a connected chain.
- A nested `if` is an `if` inside another conditional block.
- Indentation defines the block.
- Statements outside the block continue to run after the conditional.
