## 2. Lists

### Why do we need a list?

Suppose a school has 1,000 students. Creating a separate variable for every student's marks would be inconvenient. A list lets one variable hold multiple values in an ordered sequence.

A Python list:
- Is written using square brackets: `[]`.
- Holds multiple values.
- Preserves order.
- Supports indexing and slicing.
- Can contain different data types in the same list.
- Is mutable: items can be added, removed or replaced.

````python
student_marks = [80, 90, 100]
print(student_marks)

patient_info = ["Ravi", 101, 98.6, "+91-9876543210"]
print(patient_info)
print(type(patient_info))
print(type(patient_info[0]))
print(type(patient_info[1]))
print(type(patient_info[2]))
````

### 2.1 Accessing list elements: indexing

Indexing starts at `0` from the left. Negative indexing starts at `-1` from the right.

````python
student_info = ["abhinandan", 101, "Data science", "3.0"]

print(student_info[0])    # first element
print(student_info[3])    # fourth element
print(student_info[-1])   # last element

target_index = len(student_info) - 1
print(student_info[target_index])
````

### 2.2 `len()` and `type()`

- `len(collection)` returns the number of elements.
- `type(value)` tells you the object's type.

````python
print(type(student_info))
print(type(student_info[0]))
print(type(student_info[1]))
print(len(student_info))
print(student_info)
````

### 2.3 Adding elements with `append()`

`append(value)` adds one element to the end of the list. It changes the existing list.

````python
student_info = ["abhinandan", 101, "Data science", "3.0"]

student_info.append("phone-number")
print(student_info)

student_info.append("another-value")
print(student_info)
````

### 2.4 Removing elements with `pop()` and `remove()`

- `pop(index)` removes the element at the given index and returns it.
- If the index is omitted, `pop()` removes the last element.
- `remove(value)` removes the first matching value.

````python
student_info = ["abhinandan", 101, "Data science", "3.0", "phone-number"]

removed_item = student_info.pop(-1)
print("Removed:", removed_item)
print(student_info)

student_info.remove("3.0")
print(student_info)

student_info.remove("Data science")
print(student_info)
````

### 2.5 Inserting elements with `insert()`

Syntax: `list_name.insert(index, value)`

The new value is inserted at the requested position. Existing elements at that position and after it shift to the right.

````python
student_info = ["abhinandan", 101, "4.0", "phone-number"]

student_info.insert(2, "Data science")
print(student_info)

student_info.insert(3, "version")
print(student_info)
````

### 2.6 Updating an existing element

Use its index and assignment. The old value at that index is replaced.

````python
student_info = ["abhinandan", 101, "Data science", "3.0", "phone-number"]

student_info[-1] = "+1 515-67-890"
print(student_info)
````

## 3. List slicing

Slicing extracts a portion of a sequence. The same slicing idea is not exclusive to lists; it can also be used with other sequence types such as strings and tuples.

General syntax:

```python
sequence[start:stop:step]
```

- `start`: index where slicing begins (included).
- `stop`: index where slicing stops (excluded).
- `step`: how many positions to move each time.
- If omitted, defaults depend on the direction: normally start at the beginning, stop at the end, and step by `1`.

````python
students = ["student_1", "student_2", "student_3", "student_4", "student_5"]

print(students[0:3])     # indexes 0, 1, 2
print(students[-1:3])    # empty: default direction is left to right
print(students[-3:-1])   # indexes -3 and -2
print(students[-1:-3])   # empty: start/stop do not match the positive direction
````

### 3.1 Understanding the step

For a list with five elements, the indexes are:

| Element | 1 | 2 | 3 | 4 | 5 |
|---|---:|---:|---:|---:|---:|
| Positive index | 0 | 1 | 2 | 3 | 4 |
| Negative index | -5 | -4 | -3 | -2 | -1 |

A positive step moves left to right. A negative step moves right to left. The stop index remains excluded.

````python
a = [1, 2, 3, 4, 5]

print(a[0:3])       # [1, 2, 3]
print(a[0:3:2])     # [1, 3]
print(a[-1:3:-1])   # [5, 4]
print(a[-1:1:-1])   # [5, 4, 3]
print(a[-1:1:-2])   # [5, 3]
print(a[1:4])       # [2, 3, 4]
````

### 3.2 Reversing a list

These expressions move through the list in reverse order:
- `a[::-1]` is the concise reverse-slice idiom.
- `a[-1::-1]` starts explicitly at the last element and moves backwards.
- `a[-1:0:-1]` stops before index `0`, so it does **not** include the first element.

The instructor noted a preference for straightforward expressions that are easy for another reader to understand.

````python
a = [1, 2, 3, 4, 5]

print(a[-1:0:-1])   # [5, 4, 3, 2]
print(a[::-1])      # [5, 4, 3, 2, 1]
print(a[::1])       # [1, 2, 3, 4, 5]
print(a[-1::-1])    # [5, 4, 3, 2, 1]
````

### 3.3 Nested lists

A list can contain another list. Use another index to access an element inside the nested list.

````python
a = [1, 2, 3, 4, [1, 2, 3]]

print(a[-1])       # the inner list
print(a[-2])       # 4
print(a[-1][2])    # third item inside the inner list
````

## 4. Dictionaries

A dictionary stores data as **key-value pairs**. It is useful when values need meaningful labels/keys rather than only numeric positions.

Syntax:

```python
dictionary_name = {
    key: value,
    key: value
}
```

In the class example, a student ID is used as a key, and the student's details are stored as the value. Values can themselves be lists or other collection objects.

````python
students = {
    101: ["Monal", "abc@gmail.com", 80],
    102: ["Richard", "abc@gmail.com", 80],
    "103": ["Rishabh", "abc@gmail.com", 80],
    101: ["Mahesh", "abc@gmail.com", 80],
}

print(students)
````

### Important: duplicate keys

Dictionary keys must be unique. In the example above, key `101` appears twice. The later value replaces the earlier value, so the dictionary retains the value for `"Mahesh"` under key `101`.

Also note that integer key `103` and string key `"103"` are different keys.

````python
students = {
    101: ["Monal", "abc@gmail.com", 80],
    102: ["Richard", "abc@gmail.com", 80],
    "103": ["Rishabh", "abc@gmail.com", 80],
    101: ["Mahesh", "abc@gmail.com", 80],
}

print(students[101])
print(students["103"])
print(103 in students)
print("103" in students)
````

### 4.1 Accessing values: square brackets and `get()`

- `dictionary[key]` returns the value for that key. If the key is missing, it raises `KeyError`.
- `dictionary.get(key, default)` returns the value if present; otherwise it returns the supplied default (or `None` if no default is supplied).

````python
students = {
    101: ["Mahesh", "abc@gmail.com", 80],
    102: ["Richard", "abc@gmail.com", 80],
}

print(students[102])
print(students.get(102, "Value not present"))

# Missing key:
print(students.get(105, "Value not present"))

# Uncomment to see the KeyError:
# print(students[105])
````

### 4.2 Getting all keys, values and key-value pairs

- `keys()` gives the dictionary's keys.
- `values()` gives its values.
- `items()` gives key-value pairs.

These methods return dictionary view objects. The class discussed `items()` as especially useful when working with dictionaries in loops later.

````python
students = {
    101: ["Mahesh", "abc@gmail.com", 80],
    102: ["Richard", "abc@gmail.com", 80],
}

print(students.keys())
print(students.values())
print(students.items())
````

### 4.3 Adding and updating values using assignment

Syntax: `dictionary[key] = value`

Python checks whether the key already exists:
- If it exists, the value is replaced.
- If it does not exist, a new key-value pair is added.

````python
students = {
    101: ["Mahesh", "abc@gmail.com", 80],
    102: ["Richard", "abc@gmail.com", 80],
}

# New key: adds an entry
students[104] = ["shyamu", "abc@gmail.com", 80]
print(students)

# Existing key: replaces the previous value
students[104] = []
print(students)
````

### 4.4 Removing dictionary entries: `pop()` and `del`

- `dictionary.pop(key)` removes the key and returns its value.
- `del dictionary[key]` deletes the specified key-value pair.
- `del` is a Python statement, not a dictionary method.

````python
students = {
    101: ["Mahesh", "abc@gmail.com", 80],
    103: ["Rishabh", "abc@gmail.com", 80],
    104: ["shyamu", "abc@gmail.com", 80],
}

removed_value = students.pop(104)
print("Removed:", removed_value)
print(students)

del students[103]
print(students)
````

### 4.5 Adding/updating multiple entries with `update()`

`update()` accepts another mapping (or suitable key-value pairs) and adds or updates entries.

````python
students = {
    101: ["Mahesh", "abc@gmail.com", 80],
    102: ["Richard", "abc@gmail.com", 80],
}

students.update({
    103: ["Rishabh", "abc@gmail.com", 80],
    104: ["shyamu", "abc@gmail.com", 80],
})

print(students)
````

## 5. Sets

A set is a collection that keeps **unique** elements. Duplicate values are removed automatically.

A set:
- Can be created with curly braces containing values, or with `set()`.
- Does not preserve a user-controlled positional order for indexing.
- Does not support ordinary positional indexing or slicing.
- Supports adding elements with `add()`.

````python
colors = set()
print(type(colors))

colors.add("blue")
colors.add("blue")   # duplicate: no second copy is added
colors.add("green")

print(colors)
````

````python
num = {1, 2, 3, 4, 5, 5, 5, 5, 5, 6}
print(num)
print(type(num))
````

### 5.1 Converting a list to a set to remove duplicates

This is a common quick way to obtain unique values from a list. The set's displayed order should not be relied upon.

````python
ids = [1, 1, 2, 3, 4, 5, 6, 6, 12, 2, 3, 4, 5, 6, 6, 4,
       3, 3, 45, 45, 3, 3, 3, 3, 6, 7, 7, 5, 6, 6, 7, 89, 8, 9]

unique_ids = set(ids)
print(unique_ids)
print(type(unique_ids))
````

## 6. Tuples

A tuple is an ordered collection used when the collection should remain fixed after creation.

A tuple:
- Is commonly written with parentheses: `()`.
- Can contain multiple values and different data types.
- Supports indexing and slicing.
- Is immutable: its elements cannot be reassigned, appended, removed or popped.
- Can also be created using the `tuple()` constructor.

````python
# Method 1: tuple literal
week_days = ("M", "Tu", "Wed")
print(week_days)
print(type(week_days))

# Method 2: tuple constructor
tuple_temp = tuple()
print(tuple_temp)
print(type(tuple_temp))
````

### 6.1 Tuple values and immutability

Use a tuple for fixed data, such as a set of weekday labels in this example. The tuple itself cannot be changed after creation.

````python
week_days = ("Monday", "Tuesday", "Wednesday")

print(week_days[0])
print(week_days[-1])
print(week_days[0:2])

# These operations are not supported:
# week_days.append("Thursday")
# week_days[0] = "Sunday"
# week_days.pop()
````

### 6.2 A tuple can contain different data types

Immutability applies to the tuple's element slots. A tuple may contain numbers, strings, lists or dictionaries. (A mutable object nested inside a tuple may still be mutable in its own right; the tuple's references cannot be reassigned.)

````python
test_tuple = (10, "Python", 3.14, [1, 2, 3], {"id": 101})
print(test_tuple)
print(type(test_tuple))
````

## 7. Comparison of the four collection types

| Feature | List | Dictionary | Set | Tuple |
|---|---|---|---|---|
| Main purpose | Ordered sequence of values | Key-value mapping | Unique values | Fixed sequence of values |
| Syntax | `[1, 2]` | `{"a": 1}` | `{1, 2}` or `set()` | `(1, 2)` |
| Duplicates | Allowed | Keys unique | Duplicate elements removed | Allowed |
| Access | Index | Key | Membership / iteration | Index |
| Mutable? | Yes | Yes | Yes | No (tuple structure) |
| Ordered? | Yes | Insertion order maintained in modern Python | No positional order | Yes |

**Remember:** `{}` creates an empty dictionary, not an empty set. Use `set()` to create an empty set.

````python
empty_dict = {}
empty_set = set()

print(type(empty_dict))
print(type(empty_set))
````
