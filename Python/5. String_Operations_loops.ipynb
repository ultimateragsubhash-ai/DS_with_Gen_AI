## 2. Strings are sequences of characters

A string can be understood as a sequence of characters. That is why many sequence operations also work on strings.

A character is not limited to an alphabet. Spaces and punctuation marks are characters too, and they are included when counting the length of a string.

````python
values = [1, 2, 3, 4, 5]
print(len(values))

name = "Data Science"
print(len(name))  # The space is counted as a character.
````

### 2.1 Indexing and slicing strings

Because a string is a sequence, indexing and slicing work in the same way as they do with lists.

- Positive indexing begins at `0`.
- Negative indexing begins at `-1` from the right.
- In a slice, the start is included and the stop is excluded.
- `[::-1]` traverses the string in reverse.

````python
name = "Data Science"

print(name[0])       # D
print(name[1])       # a
print(name[0:5])     # Data 
print(name[::-1])    # ecneicS ataD
````

### 2.2 Concatenation with `+`

The `+` operator joins strings. The operation produces a new string; it does not automatically change the original variable.

There are two ways to store the result:

**Option A — use a new variable**
- Advantage: the original value of `name` is retained.
- Disadvantage: another variable is introduced.

**Option B — assign back to the same variable**
- Advantage: the same variable is reused.
- Disadvantage: the original value previously held by `name` is replaced.

````python
name = "Data Science"

new_name = name + name
print("Original:", name)
print("New value:", new_name)

# Reassignment example — run separately if you want to replace name:
# name = name + name
# print(name)
````

### 2.3 Repeating strings with `*`

Multiplying a string by an integer repeats the string that many times. It is repetition, not numeric multiplication.

````python
name = "Data Science"

print(name * 3)
print("-" * 10)
````

### 2.4 Membership operators: `in` and `not in`

Membership checks whether a substring (a smaller sequence of characters) exists inside a larger string. The result is a Boolean: `True` or `False`.

String membership is case-sensitive. For example, `"ATA"` and `"ata"` are different sequences.

````python
name = "Data Science"

print("Data" in name)
print("ata" in name)
print("ata" not in name)
print("ATA" in name)  # Case differs, so this is False.
````

## 3. Built-in string methods

String methods return results; the examples below print the returned values without assigning them back to the original variable. Strings are immutable, so these methods do not modify the original string in place.

### 3.1 Case conversion: `lower()`, `upper()`, `capitalize()` and `title()`

- `lower()` converts letters to lowercase.
- `upper()` converts letters to uppercase.
- `capitalize()` makes the first character uppercase and the remaining characters lowercase.
- `title()` capitalizes the beginning of each word.

````python
name_2 = "ULTImate dATA Science"

print(name_2.lower())
print(name_2.upper())
print(name_2.capitalize())
print(name_2.title())

# The original value is still unchanged:
print(name_2)
````

### 3.2 Counting occurrences with `count()`

`count(substring)` returns the number of non-overlapping occurrences of the specified substring.

The class demonstrates converting the text to a common case before counting, so uppercase and lowercase versions can be counted together without reassigning the original variable.

````python
name_2 = "ULTImate dATA Science"

print(name_2.count("A"))
print(name_2.lower().count("a"))
print(name_2.upper().count("A"))
````

### 3.3 Searching with `find()`

`find(substring)` returns the index of the first match. If the substring is not found, it returns `-1`.

The search is case-sensitive unless you normalize the case first.

````python
name_2 = "ULTImate dATA Science"

print(name_2.find("ULT"))
print(name_2.find("Sci"))
print(name_2.find("Forward deploy engineer"))  # -1 when not found
````

### 3.4 Checking the beginning and end: `startswith()` and `endswith()`

These methods return Boolean values:
- `startswith(prefix)` checks whether the string begins with the given text.
- `endswith(suffix)` checks whether the string ends with the given text.

They are case-sensitive.

````python
name_2 = "ULTImate dATA Science"

print(name_2.startswith("a"))
print(name_2.startswith("u"))
print(name_2.startswith("ULT"))
print(name_2.endswith("e"))
````

### 3.5 Cleaning and replacing text: `strip()` and `replace()`

- `strip()` removes whitespace from both ends of a string.
- `lstrip()` removes whitespace from the left end.
- `rstrip()` removes whitespace from the right end.
- `replace(old, new)` returns a string in which occurrences of `old` are replaced by `new`.

These operations return new strings; assign the result if you want to retain it.

````python
new_text_1 = " hello how are you! "

print(new_text_1.strip())
print(new_text_1.lstrip())
print(new_text_1.rstrip())

cleaned_text = new_text_1.strip()
without_exclamation = cleaned_text.replace("!", "")
print(without_exclamation)
````

### 3.6 Empty strings

An empty string contains zero characters. It is still a string.

````python
a = ""
print(a)
print(type(a))
print(len(a))
````

## 4. Splitting and joining strings

### 4.1 `split()`

`split()` breaks a string into a list of smaller strings.

- With no argument, `split()` separates on runs of whitespace.
- Passing a separator, such as `","`, tells Python what delimiter to split on.

````python
new_text_2 = "hello how are you"
word_list_1 = new_text_2.split()
print(word_list_1)

new_text_3 = "hello, how are you"
word_list_2 = new_text_3.split(",")
print(word_list_2)
````

### 4.2 `join()`

`join()` combines an iterable of strings into one string, placing the chosen separator between elements.

Read the syntax as: `separator.join(iterable_of_strings)`.

The separator is placed **between** items; it is not added before the first item or after the last item.

````python
word_list_1 = ["hello", "how", "are", "you"]

print(" ".join(word_list_1))
print(",".join(word_list_1))
print("-".join(word_list_1))
````

**Important type requirement:** the items passed to `join()` must be strings. The original class notebook includes a list containing integers, a string and a float, followed by a `join()` call. That call raises `TypeError` because not all items are strings. Convert non-string items with `str()` first.

````python
word_list_3 = [1, 2, 3, "hello", 1.20]

# This raises TypeError because the list contains non-string values:
# print(" ".join(word_list_3))

joined = " ".join(str(item) for item in word_list_3)
print(joined)
````

## 5. String formatting

String formatting is useful when output needs to combine fixed text with values stored in variables. The class demonstrates three approaches.

### 5.1 Concatenation with `str()`

When joining strings with `+`, convert numeric values to strings first.

````python
table_of = 2
multiplied_by = 4
answer = table_of * multiplied_by

string_representation = (
    str(table_of) + " X " + str(multiplied_by) + " = " + str(answer)
)
print(string_representation)
````

### 5.2 Passing values as separate `print()` arguments

`print()` can accept multiple arguments and place spaces between them by default. This avoids manually concatenating each value.

````python
table_of = 2
multiplied_by = 4
answer = table_of * multiplied_by

print(table_of, "X", multiplied_by, "=", answer)
````

### 5.3 f-strings

An f-string is prefixed with `f` and allows expressions or variable names inside curly braces.

The instructor's suggested beginner approach:
1. First write the exact output format you want.
2. Identify which parts should come from variables.
3. Replace those fixed values with `{variable_name}`.
4. Dry-run the expression to see which value each placeholder contributes.

The class notebook notes that f-strings are available from Python 3.6 onward.

````python
table_of = 2
multiplied_by = 4
answer = table_of * multiplied_by

string_representation = f"{table_of} X {multiplied_by} = {answer}"
print(string_representation)

a = 12
print(f"{a}")
print(f"{a}-hello")
````

### 5.4 The `.format()` method

The `.format()` method fills `{}` placeholders from the arguments passed to it.

````python
table_of = 2
multiplied_by = 4
answer = table_of * multiplied_by

string_representation = "{} X {} = {}".format(
    table_of, multiplied_by, answer
)
print(string_representation)
````

### 5.5 Comparison of the three approaches

| Approach | Example idea | Key point |
|---|---|---|
| Concatenation | `str(a) + " X " + str(b)` | Convert non-string values to strings |
| `print()` arguments | `print(a, "X", b)` | `print()` separates arguments with spaces by default |
| f-string | `f"{a} X {b}"` | Variables appear inside `{}` |
| `.format()` | `"{} X {}".format(a, b)` | Placeholder positions match argument order |

The class also shows that an f-string can contain only literal text; in that case it does not add dynamic behavior.

````python
string_representation_2 = f"hello how are you"
print(string_representation_2)
````

## 7. `range()` — generating a sequence of integers

The session moves from strings into loops. Before using a `for` loop, understand `range()`.

General form:

```python
range(start, stop, step)
```

- `start` is included.
- `stop` is excluded.
- `step` controls the increment/decrement.
- If omitted, `start` defaults to `0` and `step` defaults to `1`.
- The `stop` value is always excluded.

````python
print(list(range(0, 10)))
print(list(range(1, 6)))
print(list(range(2, 11, 2)))
print(list(range(10, 0, -2)))
````

### 7.1 Dry-run a `range()`

For `range(2, 11, 2)`, begin at `2` and add `2` each time:

| Iteration | Value |
|---:|---:|
| 1 | 2 |
| 2 | 4 |
| 3 | 6 |
| 4 | 8 |
| 5 | 10 |
| Next | 12 — stop because 12 is not less than 11 |

So the generated values are `2, 4, 6, 8, 10`. The end value `11` is not included.

````python
for i in range(2, 11, 2):
    print(i)
````

## 8. `for` loops

A `for` loop repeats a block of code once for each item produced by an iterable, such as a list or a `range`.

General form:

```python
for variable in iterable:
    # indented statements execute for each item
```

The loop provides the next value to the loop variable. The programmer decides what to do with that value inside the loop.

````python
for i in range(1, 6):
    print(i)
````

### 8.1 Looping over a list

A `for` loop can iterate over the values in a list directly. The supplied `multiply_by_itself.py` script uses this to print a message for each item.

````python
for i in [1, 2, 3, 4, "data"]:
    print("--" * 5)
    print(f"Pouring water in glass no : {i}")
    print("--" * 5)
````

### 8.2 A loop's indentation matters

Statements indented under the `for` line belong to the loop body and run on each iteration. Statements aligned with the `for` line run after the loop finishes.

````python
print("Before loop")

for i in range(1, 4):
    print("Inside loop:", i)

print("After loop")
````

## 9. Multiplication tables

### 9.1 Repeated code without a loop

The supplied `multiplication_table.py` demonstrates the repetitive approach: calculate the product, format the output and print it, then increase the multiplier and repeat the same steps.

This works, but the same block of code is repeated many times. A loop avoids that duplication.

````python
table_of = 2
multiplied_by = 1
answer = table_of * multiplied_by
print(f"{table_of} X {multiplied_by} = {answer}")

multiplied_by += 1
answer = table_of * multiplied_by
print(f"{table_of} X {multiplied_by} = {answer}")
````

### 9.2 Multiplication table using a `for` loop

The loop supplies the multiplier. The f-string displays the table line.

````python
for i in range(1, 11):
    print(f"2 x {i} = {2 * i}")
    print("--" * 5)
````

### 9.3 Using a variable for the table number

Instead of hardcoding `2` in the expression, store the table number in a variable. This makes the program easier to reuse for another table.

````python
table_of = 7

for i in range(1, 11):
    print(f"{table_of} x {i} = {table_of * i}")
````

### 9.4 The class challenge: multipliers 1, 2, 4, 6, 8 and 10

The instructor asked for the multiplication table of 2 using only `range()` and a `for` loop, with these multipliers: `1, 2, 4, 6, 8, 10`.

One solution uses two loops: one for the special multiplier `1`, and one for the even multipliers.

````python
# Method 1 — separate the special value 1 from the even sequence
for i in range(1, 2):
    print(f"2 x {i} = {2 * i}")

for i in range(2, 11, 2):
    print(f"2 x {i} = {2 * i}")
````

A second solution uses one range and a condition. `or` allows the body to run when either the multiplier is `1` or the multiplier is divisible by `2`.

````python
# Method 2 — one loop with a condition
for i in range(1, 11):
    if i == 1 or i % 2 == 0:
        print(f"2 x {i} = {2 * i}")
````

### 9.5 Dry-run Method 2

The loop receives the values `1` through `10`. For each value, check the condition:

`i == 1 or i % 2 == 0`

| `i` | `i == 1` | `i % 2 == 0` | Print? |

| 2 | False | True | Yes |
| 3 | False | False | No |
| 4 | False | True | Yes |
| 5 | False | False | No |
| 6 | False | True | Yes |
| 7 | False | False | No |
| 8 | False | True | Yes |
| 9 | False | False | No |

The result is the required set of multipliers: `1, 2, 4, 6, 8, 10`.

The instructor emphasizes that the loop's job is to provide values; the programmer's job is to reason about how to use them.

## 10. Modulus `%` and even/odd logic

The modulus operator `%` returns the remainder after division.

- If `i % 2 == 0`, the number is divisible by 2 with no remainder: it is even.
- If `i % 2 != 0`, the number has a remainder when divided by 2: it is odd.

This logic is used in the class's even/odd examples.

````python
for i in range(1, 11):
    if i % 2 == 0:
        print(f"Number {i} is even")
    else:
        print(f"Number {i} is odd")
````

### 10.1 Store even and odd values in separate lists

The supplied `even_odd_number.py` first prints whether each value is even or odd. It then demonstrates another approach: create two empty lists and append each value to the appropriate list. Print both lists after the loop.

````python
n = 100
even = []
odd = []

for i in range(-100, n):
    if i % 2 == 0:
        even.append(i)
    else:
        odd.append(i)

print(f"Even numbers : {even}")
print(f"Odd numbers : {odd}")
````

### 10.2 Why use `append()` here?

The condition decides which list should receive the current number:
- Even number → `even.append(i)`
- Odd number → `odd.append(i)`

The loop processes the values one at a time. When the loop ends, both lists contain the values collected during all iterations.
