## 1. Variables and Python objects

A **variable name** is a name that lets us refer to an object in a Python program.

The class used the expression:

```python
a = 3
```

A useful way to think about it is:

1. Python evaluates the value `3`.
2. Python creates or reuses an integer object representing that value.
3. The name `a` is bound to that object.
4. When we use `a`, Python looks up the object associated with that name.

The class described this using a high-level RAM / operating-system illustration: Python requests memory, the operating system manages the memory, and Python keeps a way to refer to the stored object.

**Important precision:** Python's language model is best described in terms of *names and objects*. A Python variable is not literally a box that always contains the value, and the operating system does not necessarily hand Python a simple address for every assignment. The RAM illustration is a teaching model, not a complete description of Python's implementation.

````python
a = 3
print(a)
print(type(a))
````

**Expected output**

```text
3
<class 'int'>
```

The value `3` is an integer (`int`). The name `a` refers to it.

## 2. `id()` and object identity

Python's built-in `id()` function returns an integer that identifies an object during that object's lifetime.

The class used `id()` to make the idea of identity visible. Run the next cells and compare the results on your machine.

````python
a = 3
print(id(a))
````

````python
b = 30
print(id(b))
````

The two calls show identity numbers for the objects referred to by `a` and `b`. They will generally be different here because the values are different objects.

### Two names referring to the same object

````python
person_a = 0
person_b = person_a

print(person_a, id(person_a))
print(person_b, id(person_b))
print(person_a is person_b)
````

Both names refer to the same object in this example, so their `id()` values will match and `is` will be `True`.

**Remember:**

- `id(obj)` asks for the identity of an object.
- `is` checks whether two references point to the same object.
- `==` checks whether two objects have equal values.

**Implementation note:** In CPython, `id()` is commonly related to an object's memory address, but Python code should not depend on `id()` being a raw memory address. The portable meaning is object identity.

````python
first = [10, 20]
second = [10, 20]

print(first == second)  # Same contents?
print(first is second)  # Same object?
print(id(first), id(second))
````

Expected result: `first == second` is `True`, but `first is second` is `False`. They are separate list objects with equal contents.

## 3. Variable naming rules

The class covered these naming rules:

1. A name can contain **letters, digits, and underscores**.
2. A name **cannot begin with a digit**.
3. Names are **case-sensitive** (`name`, `Name`, and `NAME` are different).
4. Python keywords must not be used as variable names (`if`, `else`, `for`, etc.).
5. Prefer meaningful names that communicate what the value represents.

### Valid and invalid examples

````python
# Valid names
_data_0 = 0
data1 = 1
temperature_c = 34.6
person_a_score = 0

# These are invalid identifiers; uncomment one at a time to see SyntaxError:
# 0_data = 10
# first-name = "Subhash"
# class = "Python"
````

### Naming examples

| Name | Valid? | Reason |
|---|---:|---|
| `data_0` | Yes | Letters, underscore, digit; starts with a letter |
| `_data_0` | Yes | Leading underscore is allowed |
| `temperature_c` | Yes | Meaningful and valid |
| `0_data` | No | Starts with a digit |
| `first-name` | No | Hyphen is not allowed in an identifier |
| `class` | No | Reserved Python keyword |
| `name` and `Name` | Both valid | Python is case-sensitive |

### Meaningful names

Prefer:

```python
height_cm = 175
temporary_value = 0
temperature_c = 34.6
```

over names such as:

```python
a = 175
b = 0
c = 34.6
```

Short names can be fine for small mathematical examples, but descriptive names make longer programs easier for other developers to understand.

You can check whether a string is a valid identifier and whether it is a keyword:

````python
import keyword

names = ["data_0", "0_data", "temperature_c", "class", "print"]

for name in names:
    print(name, "identifier:", name.isidentifier(),
          "| keyword:", keyword.iskeyword(name))
````

## 4. Dynamic typing

Python is dynamically typed: a name can be rebound to objects of different types while a program runs.

The class demonstrated this with `x = 100` followed by `x = "hello world"`.

````python
x = 100
print(x)
print(id(x))
print(type(x))

x = "hello world"
print(x)
print(id(x))
print(type(x))
````

The first binding refers to an integer. The second assignment rebinds `x` to a string object.

**Key point:** Python does not permanently declare `x` as an integer. The type belongs to the object, not permanently to the variable name.

Useful built-in types introduced in the class:

| Kind of data | Python type | Example |
|---|---|---|
| Whole number | `int` | `100` |
| Decimal number | `float` | `34.6` |
| Text | `str` | `"hello world"` |
| Boolean | `bool` | `True`, `False` |

````python
count = 100
temperature_c = 34.6
student_name = "Vaibhav"
is_active = True

print(type(count))
print(type(temperature_c))
print(type(student_name))
print(type(is_active))
````

## 5. Assignment patterns and `print()`

Python supports several assignment patterns. The class demonstrated each of the following.

### 5.1 Assign one value to one name

````python
score = 0
print(score)
````

### 5.2 Assign the same value to multiple names (chained assignment)

````python
person_a = person_b = person_c = person_d = 0

print(person_a, person_b, person_c, person_d)
print(id(person_a), id(person_b), id(person_c), id(person_d))
````

All four names are assigned the same object in this example. This is why the `id()` values match.

This is different from making four independent copies. For immutable integers, that distinction is usually harmless; with mutable objects such as lists, chained assignment can be important:

````python
# All three names refer to the same list object.
list_a = list_b = list_c = []

list_a.append("hello")
print(list_a)
print(list_b)
print(list_c)
print(list_a is list_b)
````

### 5.3 Assign different values to multiple names (unpacking)

````python
person_a, person_b, person_c, person_d = 0, 10, 20, 30

print(person_a, person_b, person_c, person_d)
````

Python matches values from left to right:

- `0` → `person_a`
- `10` → `person_b`
- `20` → `person_c`
- `30` → `person_d`

The number of names and values must match for ordinary unpacking.

````python
# The same assignments written separately:
person_a = 0
person_b = 10
person_c = 20
person_d = 30

print(person_a, person_b, person_c, person_d)
````

### 5.4 Assign different types in one statement

````python
person_a, person_b, person_c, person_d = 0, 10.0, "hello", 30

print(person_a, person_b, person_c, person_d)
print(type(person_a), type(person_b), type(person_c), type(person_d))
````

### 5.5 Printing multiple values

`print()` can display multiple arguments separated by commas. It can display variables, literal values, and expressions.

````python
value_1 = 100
value_2 = 3

print(value_1, value_2, 3, "hello")
print("Addition:", value_1 + value_2)
print("Is value_1 greater?", value_1 > value_2)
````

Python evaluates the expressions first. `print()` then displays the resulting values. It is useful for checking intermediate values while debugging.

## 6. Deleting a variable name with `del`

The class demonstrated:

````python
a = 100
print(a)

del a

# Uncomment the next line to see the resulting NameError:
# print(a)
````

`del a` removes the name binding `a` from the current scope. After that, trying to access `a` raises `NameError` because that name is no longer defined.

**Important precision about memory:** `del` does not mean “immediately erase this object's bytes from RAM.” It removes a reference (a name binding). The object may remain alive if other references still point to it. When an object is no longer reachable, Python's memory management can reclaim it; the exact timing and mechanism depend on the Python implementation. The class briefly introduced this broader topic as **garbage collection**.

````python
shared = [1, 2, 3]
another_name = shared

del shared

print(another_name)  # The list still exists because another name refers to it.
````

## 9. Arithmetic operators

Use the following two values for the examples:

````python
value_1 = 100
value_2 = 3
````

| Operator | Name | Meaning | Example |
|---|---|---|---|
| `+` | Addition | Adds values | `100 + 3` |
| `-` | Subtraction | Subtracts right from left | `100 - 3` |
| `*` | Multiplication | Multiplies values | `100 * 3` |
| `/` | Division | True division | `100 / 3` |
| `//` | Floor division | Division rounded down to a whole-number boundary | `100 // 3` |
| `%` | Modulus | Remainder after division | `100 % 3` |
| `**` | Exponentiation | Raises to a power | `100 ** 3` |

### Run all arithmetic operations

````python
print("Addition:", value_1 + value_2)
print("Subtraction:", value_1 - value_2)
print("Multiplication:", value_1 * value_2)
print("Division:", value_1 / value_2)
print("Floor division:", value_1 // value_2)
print("Modulus (remainder):", value_1 % value_2)
print("Exponentiation:", value_1 ** value_2)
````

### Division `/` versus floor division `//`

````python
print(7 / 10)
print(7 // 10)
print(10 / 3)
print(10 // 3)
````

- `/` performs true division and returns a floating-point result in these examples.
- `//` performs floor division: it rounds the quotient down toward negative infinity. With two integer operands, the result is an integer.

Be careful: floor division is not simply “remove the decimal digits” for negative values.

````python
print(7 // 3)    # 2
print(-7 // 3)   # -3, because floor(-2.333...) is -3
````

### Modulus `%` — remainder

The modulus operator returns the remainder after division.

````python
print(9 % 3)    # 0: 9 divides evenly by 3
print(10 % 3)   # 1: one is left over
print(10 % 4)   # 2: two are left over
````

### Exponentiation `**`

In Python, `**` means “raise to the power.” For example, \(2^3\) means \(2 × 2 × 2 = 8\).

````python
print(2 ** 3)
print(5 ** 2)
print(10 ** 3)
````

## 10. Relational / comparison operators

Comparison operators compare two values and return a Boolean result: `True` or `False`.

| Operator | Meaning |
|---|---|
| `>` | Greater than |
| `<` | Less than |
| `>=` | Greater than or equal to |
| `<=` | Less than or equal to |
| `!=` | Not equal to |
| `==` | Equal to |

````python
value_1 = 100
value_2 = 3

print(value_1 > value_2)
print(value_1 < value_2)
print(value_1 >= value_2)
print(value_1 <= value_2)
print(value_1 != value_2)
print(value_1 == value_2)
````

### Important: `=` versus `==`

| Symbol | Purpose | Example |
|---|---|---|
| `=` | Assignment: bind a name to a value | `score = 10` |
| `==` | Comparison: test whether values are equal | `score == 10` |

````python
score = 10       # Assignment
print(score == 10)  # Comparison; returns True
print(score == 5)   # Comparison; returns False
````

## 11. Logical operators

Logical operators combine or invert Boolean conditions.

Python's three basic logical operators are:

- `and`
- `or`
- `not`

They are written in lowercase.

### 11.1 AND truth table

`A and B` is `True` only when **both** inputs are `True`.

| A | B | A and B |
|---|---|---|
| False | False | False |
| False | True | False |
| True | False | False |
| True | True | True |

### 11.2 OR truth table

`A or B` is `True` when **at least one** input is `True`.

| A | B | A or B |
|---|---|---|
| False | False | False |
| False | True | True |
| True | False | True |
| True | True | True |

### 11.3 NOT truth table

`not` reverses a Boolean value.

| A | not A |
|---|---|
| False | True |
| True | False |

### Try the logical operators in Python

````python
logic_value_1 = True
logic_value_2 = False

print("AND:", logic_value_1 and logic_value_2)
print("OR:", logic_value_1 or logic_value_2)
print("NOT logic_value_1:", not logic_value_1)
````

### Boolean data type

Python's Boolean values are `True` and `False` (capital `T` and `F`). The type is `bool`.

````python
print(type(True))
print(type(False))

# Python is case-sensitive:
# true   # NameError: Python does not define lowercase `true` by default
````

**Note:** In Python, `and` and `or` can return one of their operands rather than always returning a literal `True` or `False` when used with non-Boolean values. In this class's examples, the operands are Boolean values, so the results are Boolean.

## 12. Binary numbers — the foundation for bitwise operators

The transcript introduced binary representation to explain bitwise operations.

Computers represent data using bits. A bit has one of two values: `0` or `1`.

### 12.1 Powers of two

Each binary position represents a power of two. From right to left, the place values are:

| Binary position | Power of 2 | Decimal value |
|---:|---:|---:|
| 4th from right | \(2^3\) | 8 |
| 3rd from right | \(2^2\) | 4 |
| 2nd from right | \(2^1\) | 2 |
| Rightmost | \(2^0\) | 1 |

For a 4-bit example, the place values are **8, 4, 2, 1**.

### 12.2 Convert decimal 10 to binary

10 = 8 + 2. Select the 8 and 2 positions:

| Place value | 8 | 4 | 2 | 1 |
|---|---:|---:|---:|---:|
| Selected? | 1 | 0 | 1 | 0 |

Therefore, decimal `10` is binary `1010`.

### 12.3 Convert decimal 3 to binary

3 = 2 + 1:

| Place value | 8 | 4 | 2 | 1 |
|---|---:|---:|---:|---:|
| Selected? | 0 | 0 | 1 | 1 |

Therefore, decimal `3` is binary `0011`.

### 12.4 Range of four bits

The largest unsigned value with four bits is:

8 + 4 + 2 + 1 = 15

So four bits can represent values from 0 through 15.

````python
print(bin(10))
print(bin(3))
print(format(10, "04b"))
print(format(3, "04b"))
````

`bin()` returns a string prefixed with `0b`. `format(number, "04b")` displays a 4-bit binary representation, padding with leading zeroes.

## 13. Bitwise operators

Bitwise operators work on the binary representation of integer values. The class focused on bitwise AND (`&`) and right shift (`>>`).

### 13.1 Bitwise AND `&`

A bitwise AND compares each pair of bits. A result bit is `1` only when both corresponding input bits are `1`.

Using the class example:

```text
10 = 1010
 3 = 0011
     ----
AND  0010 = 2
```

| Bit from 10 | Bit from 3 | AND result |
|---:|---:|---:|
| 1 | 0 | 0 |
| 0 | 0 | 0 |
| 1 | 1 | 1 |
| 0 | 1 | 0 |

````python
value_b_1 = 10
value_b_2 = 3

print(value_b_1 & value_b_2)
print(format(value_b_1, "04b"))
print(format(value_b_2, "04b"))
print(format(value_b_1 & value_b_2, "04b"))
````

### 13.2 Right shift `>>`

The right-shift operator moves the bits to the right by the requested number of positions.

For non-negative integers, shifting right by one position is equivalent to floor-dividing by 2; shifting right by two positions is equivalent to floor-dividing by 4.

````python
value = 10

print(value, "->", format(value, "04b"))
print(value >> 1, "->", format(value >> 1, "04b"))
print(value >> 2, "->", format(value >> 2, "04b"))
````

### 13.3 Other bitwise operators (reference)

The transcript primarily demonstrated `&` and `>>`. For completeness, these are common Python bitwise operators; the additional ones below are reference material rather than topics demonstrated in depth in this session.

| Operator | Name | Description |
|---|---|---|
| `&` | AND | Bit is 1 if both input bits are 1 |
| `\|` | OR | Bit is 1 if either input bit is 1 |
| `^` | XOR | Bit is 1 if input bits differ |
| `~` | NOT / inversion | Inverts integer bits (Python uses signed integer semantics) |
| `<<` | Left shift | Shifts bits to the left |
| `>>` | Right shift | Shifts bits to the right |

The instructor noted that bitwise operations are not commonly needed in everyday beginner-level application code, but they can appear in low-level work, optimization, and image-mask operations.

````python
number_a = 10
number_b = 3

print("AND:", number_a & number_b)
print("OR:", number_a | number_b)
print("XOR:", number_a ^ number_b)
print("Left shift:", number_a << 1)
print("Right shift:", number_a >> 1)
````
