## 1. Variable scope: local and global

A variable's **scope** is the region of the program where its name can be used.

The class began by comparing a variable created outside a function with a variable created inside a function.

### Example: a local variable with the same name as a global variable

In the first example, `a` is defined outside the function. Inside `dummy_func()`, the code assigns `a = 0` and increments it. That assignment creates a local variable named `a` for the function. It does not change the global `a`.

````python
a = 0

def dummy_func():
    a = 0
    a += 1
    print(a)

print(a)
dummy_func()
print(a)
````

**Expected output**
```text
0
1
0
```

Dry run:

| Step | What happens | Global `a` | Local `a` |
|---|---|---:|---:|
| Before function call | Global assignment | 0 | — |
| First `print(a)` | Prints global value | 0 | — |
| Inside function | Creates local `a = 0` | 0 | 0 |
| `a += 1` | Increments local variable | 0 | 1 |
| Function print | Prints local value | 0 | 1 |
| Function ends | Local name is no longer available outside the call | 0 | — |
| Final `print(a)` | Prints global value | 0 | — |

The key point from the demonstration: the two variables have the same name but are different bindings in different scopes.

### 1.1 What is a local variable?

A variable assigned inside a function is local to that function unless declared otherwise. It can be used within that function. Once the function call finishes, its local binding is no longer accessible from outside that function.

Function parameters are also local names. This is why parameters such as `a` and `b` can be used inside an `add()` function.

````python
def add(a, b):
    return a + b

print(add(10, 20))
````

### 1.2 Reading a global variable inside a function

The recording later demonstrates that a function can read a global variable without declaring it `global`, provided the function does not assign to that name.

The supplied `temp_7.py` example is:

```python
a = 0

def dummy_func():
    print(a)

print(a)
dummy_func()
print(a)
```

The global value is read from inside the function. The important distinction is between **reading** a global name and **assigning** to that name.

````python
a = 0

def dummy_func():
    print(a)  # Reads the global variable

print(a)
dummy_func()
print(a)
````

### 1.3 Updating a global variable with `global`

If a function needs to reassign a global variable, declare that name with `global` inside the function before assigning to it.

The class's `temp_2.py` example:

````python
a = 0

def dummy_func():
    global a
    a += 1
    print(a)

print(a)
dummy_func()
print(a)
````

**Expected output**
```text
0
1
1
```

Dry run:
1. Global `a` starts at `0`.
2. The first print displays `0`.
3. `global a` tells Python that assignments to `a` in this function refer to the global name.
4. `a += 1` changes the global value to `1`.
5. The function prints `1`.
6. The final print also displays `1`.

### A useful distinction

- Reading a global variable inside a function is allowed.
- Assigning to a name inside a function normally makes that name local.
- Use `global name` when you intend to reassign the module-level global name.

This distinction explains the `UnboundLocalError` seen in the initial example: Python treats `a` as local because the function assigns to it, but the function tries to read it before that local variable has been assigned.

````python
a = 0

def dummy_func():
    # Because this function assigns to a, Python treats a as local.
    # The right-hand side tries to read that local a before it has a value.
    a += 1

# Uncomment to observe UnboundLocalError:
# dummy_func()
````

## 2. Mutable and immutable values in functions

The recording then compared an integer with a list.

- A list is mutable: its contents can be changed in place.
- A string is immutable: its characters cannot be changed in place.
- Integers are also immutable.

This distinction helps explain why appending to a global list works without a `global` declaration, while rebinding a global integer requires `global`.

### 2.1 Mutating a global list inside a function

The class's `temp_3.py` example:

````python
a = []

def dummy_func():
    a.append(123)
    print(a)

print(a)
dummy_func()
dummy_func()
print(a)
````

**Expected output**
```text
[]
[123]
[123, 123]
[123, 123]
```

The function calls `append()` on the existing list object. It changes the object's contents; it does not reassign the name `a` to a different list. Therefore, no `global a` declaration is needed for this mutation.

### 2.2 Creating a local list with the same name

The supplied `temp_4.py` example explicitly creates a new list inside the function:

````python
a = []

def dummy_func():
    a = []
    a.append(123)
    print(a)

print(a)
dummy_func()
print(a)
````

**Expected output**
```text
[]
[123]
[]
```

The `a = []` statement inside the function creates a local list. Appending to that local list does not change the global list.

### 2.3 Strings are immutable

The class's `temp_5.py` example uses `lower()`:

```python
a = "Data"

def dummy_func(s):
    s = s.lower()
    return s

print(a)
value = dummy_func(a)
print(value)
print(id(a))
print(id(value))
```

`lower()` returns a new lowercase string. It does not modify the original string in place. The returned string is stored in `value`.

`id()` returns an identity value for an object during its lifetime. The class used it to illustrate that the original and transformed strings are different objects in this example.

````python
a = "Data"

def dummy_func(s):
    s = s.lower()
    return s

print(a)
value = dummy_func(a)
print(value)

print("id(a):", id(a))
print("id(value):", id(value))
````

### 2.4 Local variable example without changing the global name

The supplied `temp_6.py` file shows a function that assigns `a = 0` locally. The `global a` line is commented out, so the function's `a` is a local name.

````python
a = "Data"

def dummy():
    # global a
    a = 0
    print(a)

dummy()
print(a)
````

### 2.5 Summary: scope and mutability

| Example | Operation inside function | Global object/name changed? | Why |
|---|---|---|---|
| Integer with `a += 1` | Reassignment | No, unless `global a` is declared | Assignment creates a local binding by default |
| Integer with `global a` | Reassignment | Yes | `global` makes assignment target the global name |
| List with `a.append(123)` | In-place mutation | Yes, the shared list object's contents change | No rebinding of `a` |
| List with `a = []` | Rebinding | No | A new local list is created |
| String with `s = s.lower()` | Rebinding to returned string | Original string remains unchanged | Strings are immutable |

**Mental model from the class:** a variable name refers to an object. Mutating a mutable object changes that object; assigning a new value to a name changes what that name refers to.

## 3. Loop control statements

The class introduced three statements that affect execution inside a loop:

- `pass` — do nothing; it is a placeholder statement.
- `continue` — skip the remaining statements in the current iteration and move to the next iteration.
- `break` — exit the nearest enclosing loop immediately.

They are not interchangeable.

### 3.1 `pass`: a placeholder

`pass` does not skip the loop iteration. It simply performs no operation at that point, and execution continues with the next statement.

It is useful when a block is syntactically required but you have not written the logic yet, or when you intentionally want no action for a condition.

````python
for i in range(1, 6):
    if i == 3:
        pass
    print(i)
````

**Output**
```text
1
2
3
4
5
```

The print for `3` still runs because `pass` does not change control flow.

### 3.2 `continue`: skip the current iteration's remaining statements

The supplied `continue.py` first shows a loop from 1 through 10. When `i == 5`, `continue` skips the statements that follow it in that iteration. The loop then proceeds with `i == 6`.

````python
for i in range(1, 11):
    if i == 5:
        continue
    print(i)
````

**Output**
```text
1
2
3
4
6
7
8
9
10
```

Dry run for `i == 5`:
1. The loop assigns `i = 5`.
2. The condition `i == 5` is true.
3. Python executes `continue`.
4. The remaining statements in this iteration are skipped.
5. The loop advances to the next value, `6`.

### `pass` versus `continue`

The supplied script compares the two ideas:

````python
# pass: execution continues to the print statement
for i in range(1, 6):
    if i == 3:
        pass
    print("pass example:", i)

# continue: the print statement is skipped for i == 3
for i in range(1, 6):
    if i == 3:
        continue
    print("continue example:", i)
````

The output differs for the value `3`: the `pass` loop prints it, while the `continue` loop does not.

### 3.3 `continue` with odd numbers

The supplied `continue.py` file contains this example. When a number is even, `continue` skips the print statements, so only odd values reach them.

````python
for i in range(1, 31):
    if i % 2 == 0:
        continue
    print("Number:", i)
    print("Not an even number")
    print("--" * 10)
````

### 3.4 `break`: exit the loop

`break` immediately terminates the nearest enclosing loop. Execution continues with the first statement after the loop.

The supplied `break.py` example stops when the first even number is reached:

````python
for i in range(1, 31):
    if i % 2 == 0:
        break
    print("Number:", i)
    print("Not an even number")
    print("--" * 10)

print("after for loop")
````

The loop prints the information for `1`. When `i` becomes `2`, the condition is true and `break` exits the loop. The final `print("after for loop")` still runs.

### 3.5 `break` and `continue` in the same loop

The supplied `break_continue.py` example:

````python
for num in range(1, 11):
    if num == 5:
        print("Stopping loop")
        break

    if num == 2:
        print("Continue")
        continue

    print(num)
````

Dry run:
- `num == 1`: prints `1`.
- `num == 2`: prints `Continue`, then skips the final print for this iteration.
- `num == 3`: prints `3`.
- `num == 4`: prints `4`.
- `num == 5`: prints `Stopping loop`, then exits the loop.
- Values 6 through 10 are never processed.

### Comparison table

| Statement | Effect | Does the loop continue? |
|---|---|---|
| `pass` | Does nothing | Yes, with the next statement in the same iteration |
| `continue` | Skips remaining statements in current iteration | Yes, with the next iteration |
| `break` | Exits the nearest loop | No, that loop ends |

## 4. Why exception handling is needed

The class described exception handling as error handling. The motivation was that, without handling an exception, Python stops normal execution when an unhandled error occurs.

The recording first used a simple addition function:

````python
def add(val_1, val_2):
    return val_1 + val_2

print(add(100, 200))
print(add(100, "200"))  # Raises TypeError
````

The first call works because both values are numbers. The second call raises a `TypeError` because Python cannot add an integer and a string in this way.

The instructor's point: if an error occurs in a running script and is not handled, statements after the error are not executed. In a real application, developers generally need to decide what the program should do when an operation fails.

## 5. Basic `try` and `except`

- Put the operation that may fail inside `try`.
- Put the response to a failure inside `except`.
- A `try` block must be followed by at least one `except` or `finally` clause in valid Python syntax; the class focused on the `try`/`except` pair.

The original class notebook included:

````python
value = 100

try:
    value / 0
except:
    print("Program failed")
````

The `try` block attempts the division. Since the divisor is zero, Python raises `ZeroDivisionError` and transfers control to the `except` block.

The original notebook also included these expressions as examples of operations that can raise errors:

````python
# Run separately to inspect the exceptions:
# value / 0       # ZeroDivisionError
# "100" / 12      # TypeError

float("12.0")     # Valid conversion
str(12.0)         # Valid conversion
````

### Catching an exception and inspecting its message

Instead of using a bare `except`, the recording demonstrated catching an exception object with `as e`:

```python
except Exception as e:
    print("Program failed:", e)
```

This lets the program display the exception message, which can help with debugging.

````python
value = 100
value_2 = 0

try:
    value_3 = value / value_2
    print(value_3)
except Exception as e:
    print("Program failed:", e)

print("End of python script")
````

Expected behaviour: the division fails, the exception is caught, the message is printed, and then the script continues to the final print.

`Exception` is a broad base class for many ordinary runtime exceptions. Catching it broadly can be useful at an application boundary, but catching the specific exception you expect usually makes the code easier to understand and debug.

## 6. Specific exception handling

The recording then compared a generic handler with separate handlers for different exception types.

Example exception names appearing in the supplied materials:
- `ZeroDivisionError` — division by zero.
- `TypeError` — an operation receives an unsupported type combination.
- `FileNotFoundError` — the requested file path does not identify an existing file.
- `UnboundLocalError` — a local variable is referenced before it has been assigned.

You can use multiple `except` clauses. Python runs the first matching handler.

````python
value = 100
value_2 = 0

try:
    result = value / value_2
except ZeroDivisionError:
    print("Cannot divide by zero.")
except TypeError:
    print("The supplied values have incompatible types.")
````

The structure is similar to branching with `if`/`elif`: each handler represents a different failure case. The instructor noted that there can be multiple `except` clauses, depending on the possible errors the code needs to handle.

## 7. `else` and `finally`

The class explained the full structure:

```python
try:
    # Main operation that may raise an exception
except SomeException:
    # Runs if a matching exception is raised
else:
    # Runs only if the try block completes without an exception
finally:
    # Runs whether the try succeeds or an exception is handled
```

- `else` is optional. It runs only when the `try` block completes without raising an exception.
- `finally` is optional. It runs whether the `try` block succeeds or fails, and is commonly used for cleanup.
- `try` and an exception-handling path must be structured together; `except` cannot stand alone.

````python
value = 100
value_2 = 0
value_3 = 0

try:
    value_3 = value / value_2
    print(value_3)
except Exception as e:
    print("Program failed:", e)
else:
    print("Logic worked correctly.")
finally:
    print("Program cleanup is done")
````

Since `value_2` is zero, the `except` block runs. The `else` block does not run because the `try` block did not complete successfully. The `finally` block still runs.

The supplied `exception_handling_test.py` also deletes the variables in `finally`:

```python
finally:
    del value
    del value_2
    del value_3
    print("Program cleanup is done")
```

That is the specific cleanup example supplied with the class. In practical programs, `finally` is often used for releasing resources such as files or connections.

## 8. A file-reading function that can fail

The main practical example combines the previous class's file-handling topic with exception handling.

The supplied `exception_handling_final.py` has a function that:
1. Builds a path from the current working directory and a filename.
2. Opens the file.
3. Reads and strips its content.
4. Converts the content to a `float`.
5. Returns the numeric value.

The supplied text file `variable.txt` contains a single value (`100`).

````python
import os

def get_input_from_txt(file_name):
    filepath = os.path.join(os.getcwd(), file_name)
    f = open(filepath, "r+")
    content = f.read().strip()
    number = float(content)
    return number

# Example call; requires variable.txt in the working directory.
# data = get_input_from_txt("variable.txt")
# print(data)
````

### Potential failures in this function

- The file name/path may be wrong → `FileNotFoundError`.
- The content may not represent a valid number → `ValueError`.
- The file operation may fail for another reason, depending on permissions or environment.

The class focused on `FileNotFoundError` and `ZeroDivisionError` in the larger example.

## 9. Main calculation and specific exception paths

The supplied script also wraps the calculation in a function:

- Load a number from a text file.
- Ask the user for a divisor.
- Convert the input to `float`.
- Divide the file value by the user value.
- Print the result.

````python
def main_code(file_name):
    data = get_input_from_txt(file_name)
    number = float(input("Enter a number: "))
    value = data / number
    print(value)

# Interactive example:
# main_code("variable.txt")
````

### First path: the file is not found

The original script attempts to handle `FileNotFoundError` by using a default/correct file name and then running the main operation again. The recording emphasized that the fallback path itself can also fail — for example, the user may still enter zero — so that fallback needs its own exception handling.

### Second path: the file exists, but the user enters zero

If the original file exists and is read successfully, the main calculation can still raise `ZeroDivisionError` when the user enters `0`. That is a separate failure case and should have its own handler.

The teaching point was to trace the control flow one path at a time, rather than treating the entire nested structure as one undifferentiated block.

````python
file_name = "variable.txt"

try:
    main_code(file_name)

except FileNotFoundError:
    print("The requested file was not found. Trying the default file.")

    # The fallback operation can fail too, so it gets its own try/except.
    try:
        main_code("variable.txt")
    except ZeroDivisionError:
        print("The user tried to divide by zero in the fallback path.")
    except Exception as e:
        print("Fallback path failed:", e)

except ZeroDivisionError:
    print("The user tried to divide by zero.")

except Exception as e:
    print("Something else went wrong:", e)

else:
    print("Code executed successfully.")

finally:
    print("Data cleaning done.")
````

**Important note about the original supplied script:** the source file contains a nested retry path and a broad bare `except`. This notebook preserves the lesson's intended control-flow design while making the fallback path explicit and adding an exception message to the generic handler. The example remains interactive and expects `variable.txt` in the notebook's working directory for the successful path.

A production implementation should also close files reliably (for example, with `with open(...) as f`) and should decide whether retrying with the same fallback filename is meaningful for the application.

## 10. How nested exception handling works

The instructor drew the flow as separate levels:

1. The outer `try` attempts the normal file-and-calculation path.
2. If the file is missing, the outer `FileNotFoundError` handler runs.
3. That handler tries a fallback path.
4. The fallback has its own `try`/`except` because it can produce another error.
5. If the user enters zero in the fallback path, the nested `ZeroDivisionError` handler handles it.
6. If the original file exists but the user enters zero, the outer `ZeroDivisionError` handler handles that path.
7. `else` runs only when the outer `try` succeeds without an exception.
8. `finally` runs at the end regardless of success or handled failure.

### Flow sketch

```text
Outer try
├── Normal file exists and calculation succeeds
│   └── else
├── FileNotFoundError
│   └── Outer except
│       └── Nested try: fallback file + calculation
│           ├── Success
│           ├── ZeroDivisionError -> nested except
│           └── Other exception -> nested generic except
├── ZeroDivisionError -> outer except
└── Other exception -> outer generic except

finally -> runs at the end
```

The main lesson from the recording: an exception raised inside an `except` handler is not automatically caught by another `except` clause belonging to the same `try`. If the fallback logic can fail, it needs its own `try`/`except` or the exception must be allowed to propagate to a higher-level handler.

## 11. Generic exception handler: use with care

The recording demonstrated a final generic handler that catches errors not matched by the earlier specific handlers:

```python
except:
    print("Something went wrong")
```

The supplied script uses this style. It is useful to know the syntax, but a bare `except` also catches exceptions that may be better allowed to propagate (including some system-exiting exceptions).

Prefer a generic `except Exception as e` when you genuinely need a broad fallback, because it lets you inspect the error and does not catch `BaseException` subclasses such as `KeyboardInterrupt` and `SystemExit`.

````python
try:
    result = int("not-a-number")
except ValueError as e:
    print("Invalid number:", e)
except Exception as e:
    print("Unexpected error:", e)
````

| Concept | Meaning |
|---|---|

| Clause | When it runs |
