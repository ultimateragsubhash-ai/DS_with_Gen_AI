## 1. Assignment operators

The simple assignment operator `=` assigns a value to a name:

```python
a = 10
```

The class then introduced **augmented assignment operators**. They combine an operation and assignment in one shorter statement.

### 1.1 Addition assignment: `+=`

The instructor's example:

```python
b = 1
b += 2
```

This is equivalent in meaning to `b = b + 2`: Python reads the current value, adds 2, and assigns the result back to `b`.

````python
b = 1
b += 2
print(b)
````

**Expected output:** `3`

### 1.2 Dry run

The instructor emphasized doing a **dry run**: manually trace the program statement by statement and record how values change.

| Step | Statement | Value of `b` |
|---:|---|---:|
| 1 | `b = 1` | 1 |
| 2 | `b += 2` → `b = b + 2` | 1 + 2 = 3 |
| 3 | `print(b)` | Prints 3 |

Dry-running helps you understand execution order and variable updates before running the code.

### 1.3 Other augmented assignment operators

| Operator | Expanded form | Meaning |
|---|---|---|
| `+=` | `x = x + value` | Add and assign |
| `-=` | `x = x - value` | Subtract and assign |
| `*=` | `x = x * value` | Multiply and assign |
| `/=` | `x = x / value` | Divide and assign |
| `//=` | `x = x // value` | Floor-divide and assign |

The advantage discussed in class is shorter notation: the variable name does not need to be repeated on both sides.

````python
b = 1
b += 2
print("After += :", b)

b = 3
b -= 2
print("After -= :", b)

b = 3
b *= 2
print("After *= :", b)

b = 3
b /= 2
print("After /= :", b)

b = 3
b //= 2
print("After //= :", b)
````

`/=` performs true division and generally produces a float. `//=` performs floor division.

````python
b = 3
b /= 2
print(b, type(b))

b = 3
b //= 2
print(b, type(b))
````

## 4. Input with `input()`

`input()` allows a program to receive text typed by the user.

**Important class takeaway: `input()` always returns a string.** Even if the user types digits such as `12`, Python receives the text `"12"` unless the program converts it.

````python
user_input_1 = input("Enter first value: ")
user_input_2 = input("Enter second value: ")

print("User input 1:", user_input_1)
print("User input 2:", user_input_2)
print(type(user_input_1), type(user_input_2))
````

Try entering `12` and `10`. The values look numeric on screen, but their type is `str`.

### 4.1 Why does `"12" + "10"` produce `"1210"`?

The `+` operator depends on the data types:

- Number + number → arithmetic addition.
- String + string → concatenation (joining text).

````python
number_1 = 12
number_2 = 10
print(number_1 + number_2)

text_1 = "12"
text_2 = "10"
print(text_1 + text_2)
print(type(text_1 + text_2))
````

**Expected output**

```text
22
1210
<class 'str'>
```

This is why input values usually need conversion before numeric calculations.

## 6. Type casting

**Type casting** means converting a value from one data type to another.

The class used these functions:

| Function | Converts to | Example |
|---|---|---|
| `int()` | Integer | `int("12")` |
| `float()` | Floating-point number | `float("12.90")` |
| `str()` | String | `str(12.90)` |

### 6.1 Convert input strings to integers

````python
user_input_1 = input("Enter first number: ")
user_input_2 = input("Enter second number: ")

print("Before conversion:", type(user_input_1), type(user_input_2))

user_input_1 = int(user_input_1)
user_input_2 = int(user_input_2)

print("After conversion:", type(user_input_1), type(user_input_2))
print("Sum:", user_input_1 + user_input_2)
````

If you enter `12` and `10`, the final sum is `22` because both strings are converted to integers before addition.

### 6.2 Convert strings and numbers

The instructor demonstrated these conversions:

````python
print(int("12"))        # String containing an integer -> int
print(float("12.90"))   # String containing a decimal -> float
print(int(12.90))       # Float -> int
print(str(12.90))       # Float -> string
print(float(10))        # Int -> float
````

Expected values:

```text
12
12.9
12
12.9
10.0
```

A numeric float such as `12.90` is represented as `12.9`; the trailing zero does not change its numeric value. If exact textual formatting such as `12.90` matters, preserve or format it as text.

### 6.3 Why does `int("12.90")` fail?

The class specifically demonstrated that this conversion raises `ValueError`.

````python
# This raises ValueError:
# int("12.90")
````

`int()` can parse a string such as `"12"` as an integer, but it cannot directly parse `"12.90"` as an integer.

The class showed an intermediate conversion:

````python
decimal_text = "12.90"

as_float = float(decimal_text)
as_integer = int(as_float)

print(as_float)
print(as_integer)
````

This works, but converting the float `12.9` to an integer gives `12`; the fractional part is discarded (truncated toward zero).

### 6.4 Input directly converted to float

The supplied `file-1.py` uses a shorter approach:

````python
user_input_1 = float(input("Enter a number: "))
user_input_2 = float(input("Enter another number: "))

print("User input 1:", user_input_1)
print("User input 2:", user_input_2)
print("Types:", type(user_input_1), type(user_input_2))
print("Sum:", user_input_1 + user_input_2)
````

This combines input and conversion:

1. `input()` receives text.
2. `float()` converts that text to a floating-point number.
3. The converted number is assigned to the variable.

If the user enters text such as `hello`, conversion fails with `ValueError`. The input must be interpretable as a float.

## 7. Nested function calls

The instructor explained that a function can be placed inside another function.

Example:

```python
float(input("Enter a number: "))
```

Python evaluates the inner function first:

1. `input(...)` displays the prompt and receives text.
2. The returned string is passed to `float(...)`.
3. `float()` converts the string to a floating-point number.
4. The result is assigned to the variable or passed to another function.

````python
converted_number = float(input("Enter a decimal number: "))
print("You entered:", converted_number)
print("Type:", type(converted_number))
````

The same nesting idea works here:

````python
print(type(float("12.90")))
````

Python first evaluates `float("12.90")`, passes the result to `type()`, and then prints the returned type.

## 8. Combining relational and logical operators

The original notebook and Python script include this scenario:

> Take a person's age and the current time in 24-hour format. Decide whether the person meets the age and time conditions for a party.

Conditions in the example:
- Age must be at least 18.
- Current time must be between 18 and 23 inclusive.

### 8.1 Break the logic into smaller conditions

````python
age = 18
current_time_24_hr = 16

check_age = age >= 18
check_current_time_condition = (
    current_time_24_hr >= 18
    and current_time_24_hr <= 23
)

print("Age condition:", check_age)
print("Time condition:", check_current_time_condition)

can_they_party = check_age and check_current_time_condition
print("Allow in club?", can_they_party)
````

For the given values, the age condition is `True`, but the time condition is `False` because the time is 16. Therefore, the final result is `False`.

### 8.2 Dry run the conditions

| Expression | Evaluation | Result |
|---|---|---|
| `age >= 18` | `18 >= 18` | `True` |
| `current_time_24_hr >= 18` | `16 >= 18` | `False` |
| `current_time_24_hr <= 23` | `16 <= 23` | `True` |
| Time condition | `False and True` | `False` |
| Final condition | `True and False` | `False` |

### 8.3 Write the condition in one expression

````python
age = 18
current_time_24_hr = 16

can_they_party = (
    age >= 18
    and current_time_24_hr >= 18
    and current_time_24_hr <= 23
)

print("Allow in club?", can_they_party)
````

Both approaches express the same logical requirement. Storing intermediate Boolean results can make a dry run easier; the second form writes the complete condition directly.

This is a classroom logic exercise, not a real-world access-control rule.

## 9. Python virtual environments

The latter part of the class discussed how Python projects use separate environments.

A **virtual environment** is an isolated Python environment for a project. It helps keep project dependencies separate, so different projects can use different package versions without unnecessarily affecting one another.

The instructor asked learners to understand:
- What a virtual environment is.
- How to create one using Python's built-in `venv`.
- What the `uv` tool does.
- Why a tool such as `uv` can be useful when managing environments and Python versions.

### 9.1 Python's built-in `venv`

The transcript mentions the built-in Python approach as `python -m venv`.

A common Windows example:

````python
# Run in a terminal:
# python -m venv .venv
````

A common Windows PowerShell activation command:

````python
# Run in Windows PowerShell:
# .\.venv\Scripts\Activate.ps1
````

For macOS/Linux:

````python
# Run in macOS/Linux terminal:
# source .venv/bin/activate
````

### 9.2 Local and shared environment folders

During the screen-share discussion, the instructor distinguished:

- **Local/project environment:** the environment folder (often `.venv`) is inside the project folder.
- **Shared or centrally managed environment folder:** environments are stored in a separate folder, and a project activates the desired environment from there.

The exact commands and paths used during the screen share are not fully recoverable from the transcript, so the examples below are general patterns, not a verbatim copy of the instructor's terminal.

````python
# Example: project-local environment
# python -m venv .venv

# Example: environment in a separate folder
# python -m venv "C:\path\to\python-envs\project-env"
````

**Terminology clarification:** A Python virtual environment is not the same as the operating system's global Python installation. In this class, “global” was used informally to describe storing environments in a central folder rather than inside each project.

### 9.3 Why activate an existing environment again?

The class discussion clarified that you do not need to recreate an environment every time you open a terminal or restart the computer.

- The environment's files remain on disk.
- Activate it when you want to use it in a new terminal session.
- Installing packages into that environment changes that environment, not every Python project on the computer.

Creating and activating are different actions.

## 10. Python vs `uv`

An important distinction from the class:

- **Python** is the programming language and interpreter/runtime used to execute Python code.
- **`uv`** is a separate tool used to manage Python environments, Python versions, and packages/workflows.

The instructor used an analogy: the version number of one application (such as WhatsApp) is not the same as the version number of the phone's operating system. Likewise, the `uv` version is not the Python version selected for a project.

### 10.1 Check the versions

These are terminal commands:

````python
# Check the Python interpreter version
# python --version

# Check the uv tool version
# uv --version
````

The transcript gives an example where `uv --version` displayed a `0.x` version while the environment was configured to use Python 3.12. These are two different version numbers belonging to two different tools.

### 10.2 Why use `uv`?

The instructor highlighted these reasons:

- It can help create and manage environments.
- It can make it easier to choose a Python version for a project.
- It is lightweight, according to the class explanation.
- It can reduce manual work when projects or deployments require different Python versions.

The transcript contrasts built-in `venv` with `uv` and assigns learners to investigate the differences themselves.

### 10.3 Standard `uv` command examples (reference)

The following are standard command patterns added for practice; the transcript did not preserve every character of the live command clearly.

````python
# Create a virtual environment with uv, requesting Python 3.12
# uv venv --python 3.12 .venv

# Activate in Windows PowerShell
# .\.venv\Scripts\Activate.ps1

# Activate in macOS/Linux
# source .venv/bin/activate
````

**Important:** The `uv` version and the Python version used inside an environment are independent. For example, `uv` can be version `0.x` while the environment uses Python `3.12.x`.
